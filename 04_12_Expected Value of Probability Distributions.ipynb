## Expected Value of a Probability Distribution

### 1. Introduction

The **Expected Value** of a random variable is the **probability-weighted average of all possible values**.

It represents the long-run average outcome when the experiment is repeated many times.

For a discrete random variable:

$$
E[X] = \sum xP(X=x)
$$

In simple words:

> **Multiply each possible value by its probability and add the results.**


### 2. Simple Example


Consider a random variable with the following distribution:

| X | P(X) |
|---|---:|
| 1 | 0.2 |
| 2 | 0.5 |
| 3 | 0.3 |

The expected value is:

$$
E[X] = (1)(0.2) + (2)(0.5) + (3)(0.3)
$$

$$
E[X] = 0.2 + 1.0 + 0.9
$$

$$
E[X] = 2.1
$$

So the expected value is **2.1**.

This does not mean that `X` must actually become 2.1.

It means that if we repeat the experiment many times, the **long-run average** will tend toward 2.1.

```python
import numpy as np

x = np.array([1, 2, 3])
probabilities = np.array([0.2, 0.5, 0.3])

expected_value = np.sum(x * probabilities)

print("Expected Value:", expected_value)
```


### 3. Expected Value of Common Distributions


Different probability distributions have different expected values.

#### Bernoulli

For a Bernoulli Distribution:

$$
E[X] = p
$$

where `p` is the probability of success.

#### Binomial

For a Binomial Distribution:

$$
E[X] = np
$$

where:

- `n` = number of trials
- `p` = probability of success

#### Poisson

For a Poisson Distribution:

$$
E[X] = \lambda
$$

where `λ` is the average event rate.

#### Uniform

For a continuous Uniform Distribution between `a` and `b`:

$$
E[X] = \frac{a+b}{2}
$$

#### Normal

For a Normal Distribution:

$$
E[X] = \mu
$$

where `μ` is the mean.

### 4. Example 


 #### a. Binomial Distribution

Suppose:

- Number of trials = 100
- Probability of success = 0.2

Then:

$$
E[X] = np
$$

$$
E[X] = 100(0.2)
$$

$$
E[X] = 20
$$

The expected number of successes is **20**.

This means that over many repetitions, we would expect the average number of successes to be around 20.

```python
from scipy.stats import binom

n = 100
p = 0.2

expected_value = binom.mean(n, p)

print("Expected number of successes:", expected_value)
```

#### b. Poisson Distribution

Suppose a customer support center receives an average of **8 calls per hour**.

For a Poisson Distribution:

$$
E[X] = \lambda
$$

Here:

$$
\lambda = 8
$$

Therefore:

$$
E[X] = 8
$$

The expected number of calls is **8 calls per hour**.

This does not mean exactly 8 calls will arrive every hour. Some hours may have 5 calls, others 10 or 12.

It means that over many similar hours, the **average number of calls per hour will tend toward 8**.

```python
from scipy.stats import poisson

lambda_value = 8

expected_value = poisson.mean(lambda_value)

print("Expected number of calls per hour:", expected_value)
```

#### c. Normal Distribution

Suppose the heights of a population are approximately Normally distributed with:

- Mean = 170 cm
- Standard deviation = 10 cm

For a Normal Distribution:

$$
E[X] = \mu
$$

Therefore:

$$
E[X] = 170
$$

The expected height is **170 cm**.

Here, the expected value is simply the **mean of the distribution**.

It represents the center around which the values are distributed.

```python
from scipy.stats import norm

mean = 170
std = 10

expected_value = norm.mean(loc=mean, scale=std)

print("Expected height:", expected_value)
```

### 5. Business Applications

Expected Value is useful for estimating the average outcome of uncertain events.

Examples:

#### Marketing

If 1,000 customers have a 5% probability of purchasing:

$$
E[X] = 1000(0.05) = 50
$$

We expect around **50 purchases on average** over repeated similar situations.

#### Customer Support

If a call center receives an average of 20 calls per hour:

$$
E[X] = 20
$$

The expected number of calls per hour is 20.

#### Finance

Expected value can be used to estimate the average outcome of uncertain profits and losses.

### 6. ML Connection

Expected Value is an important idea in Machine Learning because many ML concepts involve **averaging outcomes, probabilities, or losses**.

#### 1. Expected Loss

In Machine Learning, a model can make different predictions and incur different losses.

Expected loss represents the **average loss weighted by the probability of different outcomes**.

The goal of many ML algorithms is to minimize expected loss.

#### 2. Classification

In probabilistic classification, a model can assign probabilities to possible classes.

For example:

- Spam → 0.8
- Not Spam → 0.2

These probabilities can be used to calculate expected outcomes and losses.

#### 3. Decision Making

Expected value can help compare decisions when outcomes are uncertain.

For example:

> Which action gives the highest expected reward?

This idea appears in areas such as reinforcement learning and probabilistic decision-making.

#### 4. Statistical Learning

Expected values are also used when defining population-level quantities such as:

- Expected error
- Expected loss
- Expected risk
- Expected prediction

The important connection is:

> **Probability distributions describe uncertainty, and Expected Value summarizes that uncertainty into an average outcome.**
