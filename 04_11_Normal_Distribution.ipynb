## Normal distribution

### 1. Introduction

The **Normal Distribution** is one of the most important continuous probability distributions in statistics and data science.

It is a **symmetric, bell-shaped distribution** where most observations are concentrated around the mean, and fewer observations occur as we move farther away from the mean.

Examples of variables that can often be approximately normally distributed include:

- Height
- Measurement errors
- Exam scores in some settings
- Manufacturing measurements
- Certain biological measurements

The shape and spread of a Normal Distribution are determined mainly by:

- **Mean (μ)** → center of the distribution
- **Standard deviation (σ)** → spread of the distribution


### 2. Intuition

### 3. Important Properties


A Normal Distribution has several important properties.

#### 1. Bell-shaped

The curve has a characteristic bell shape.

#### 2. Symmetric

The left and right sides are mirror images around the mean.

#### 3. Mean = Median = Mode

For a perfectly Normal Distribution:

$$
Mean = Median = Mode
$$

All three are located at the center.

#### 4. Values are concentrated around the mean

Most observations are close to the mean.

Fewer observations occur farther away from the mean.

#### 5. The tails extend in both directions

The curve approaches the x-axis but theoretically continues indefinitely in both directions.

### 4. Role of Mean and Standard Deviation


Two parameters determine the shape of a Normal Distribution.

#### Mean (μ)

The mean determines the **center** of the distribution.

Changing the mean shifts the entire curve left or right.

#### Standard Deviation (σ)

Standard deviation determines the **spread** of the distribution.

- Small σ → values are concentrated close to the mean
- Large σ → values are more spread out

So remember:

**Mean → Where is the distribution centered?**

**Standard deviation → How spread out is it?**

#### Normal Distribution


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean = 170
std = 10

x = np.linspace(130, 210, 1000)
y = norm.pdf(x, mean, std)

plt.plot(x, y)

plt.xlabel("Height (cm)")
plt.ylabel("Density")
plt.title("Normal Distribution")

plt.show()
```

### 5. Changing Standard Deviation




#### Effect of Standard Deviation

The standard deviation controls the spread of the curve.

#### Smaller standard deviation

Values are more concentrated around the mean.

The curve becomes **narrower and taller**.

#### Larger standard deviation

Values are more spread out.

The curve becomes **wider and shorter**.

The mean remains the center.

This is why standard deviation is an important parameter of the Normal Distribution.

A Bernoulli distribution has only two possible values:

**0 → Failure**

**1 → Success**

The height of each bar represents its probability.

### Compare different spreads

```python
mean = 0

x = np.linspace(-5, 5, 1000)

y1 = norm.pdf(x, mean, 0.5)
y2 = norm.pdf(x, mean, 1)
y3 = norm.pdf(x, mean, 2)

plt.plot(x, y1, label="σ = 0.5")
plt.plot(x, y2, label="σ = 1")
plt.plot(x, y3, label="σ = 2")

plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Effect of Standard Deviation")
plt.legend()

plt.show()
```

### 6. Why is Normal Distribution Important?

The Normal Distribution appears frequently in statistics because many measurements and aggregated effects can be approximately normal under appropriate conditions.

It is also important because many statistical methods are based on normality assumptions or use the Normal Distribution as an approximation.

It is commonly encountered in:

- Statistical inference
- Confidence intervals
- Hypothesis testing
- Measurement and experimental data
- Error analysis
- Machine learning and data science

### 7. Real-World Example

Suppose the heights of a population are approximately Normally distributed with:

Mean = 170 cm

Standard deviation = 10 cm

The distribution is centered around 170 cm.

Most observations will be relatively close to 170 cm, while extremely short or extremely tall values will be less common.

The Normal Distribution helps us understand how values are spread around the average.

Note:

Not every real-world dataset follows a Normal Distribution. We should examine the data rather than automatically assuming normality.

### 8. Business Applications

Normal Distribution can be useful in situations involving measurements or errors that are approximately symmetric and bell-shaped.

Examples:

- Manufacturing measurements
- Product dimensions
- Measurement errors
- Process variation
- Quality control
- Some customer or employee measurements

For example, a manufacturing company may examine whether the dimensions of products are centered around the desired specification and how much they vary.

### 9. ML Connection

Normal Distribution appears frequently in machine learning and statistics.

It is relevant to:

- Data analysis
- Feature distributions
- Standardization
- Gaussian-based models
- Statistical assumptions
- Error and residual analysis

Many ML algorithms do **not** require every feature to be Normally distributed.

So we should not assume:

> "Machine learning data must be Normal."

Instead, Normality is useful when it matches the data or when a particular statistical method makes that assumption.
