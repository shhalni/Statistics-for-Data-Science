## Discrete vs Continuos  Random Variable

Random variables can be classified into two main types based on the values they can take:

1. **Discrete Random Variables**
2. **Continuous Random Variables**

The main difference is:

- **Discrete → Countable values**
- **Continuous → Any value within a range**

Understanding this distinction is important because different probability functions are used for each type.

### Discrete Random variable

#### Intro
A **discrete random variable** can take a finite or countably infinite number of distinct values.

The possible values can be **counted individually**.

Examples:

- Number of customers
- Number of defective products
- Number of heads in coin tosses
- Number of phone calls
- Number of purchases

For example, if:

$$
X = \text{Number of customers entering a store}
$$

then possible values could be:

$$
X \in \{0,1,2,3,4,\ldots\}
$$

We can count each possible value.

#### Key Idea

> **If you can count the possible values one by one, the random variable is discrete.**

#### Intuition

Imagine counting the number of customers entering a shop.

You could have:

$$
0,\ 1,\ 2,\ 3,\ 4,\ldots
$$

But you cannot have:

$$
2.5 \text{ customers}
$$

A customer is a countable object.

Therefore, the number of customers is a **discrete random variable**.

Other examples:

- 5 defective products
- 12 website clicks
- 3 children
- 20 customer complaints

All of these are counts.

#### Think:

**Counting → Discrete**

### Continuos Random variable

#### Introduction

A **continuous random variable** can take any value within a given range.

There can be infinitely many possible values between two points.

Examples:

- Height
- Weight
- Temperature
- Time
- Distance
- Blood pressure

For example:

$$
X = \text{Height of a person}
$$

The height could be:

$$
170
$$

or:

$$
170.1
$$

or:

$$
170.15
$$

or:

$$
170.1537
$$

and so on.

There are infinitely many possible values within a range.

#### Key Idea

> **If the variable can take any value within a range, it is continuous.**

#### Intuition

Imagine measuring someone's height.

Suppose the height is between 170 cm and 171 cm.

Possible values include:

$$
170,\ 170.1,\ 170.01,\ 170.001,\ 170.0001,\ldots
$$

There is no smallest possible difference between two measurements in the theoretical mathematical model.

Therefore, height is treated as a **continuous random variable**.

#### Think:

**Measuring → Continuous**

###  Discrete vs Continuous



| Feature | Discrete | Continuous |
|---|---|---|
| Values | Countable | Infinitely many within a range |
| Usually represents | Counts | Measurements |
| Example | Number of customers | Customer waiting time |
| Can have decimals? | Usually no | Yes |
| Example values | 0, 1, 2, 3 | 2.1, 2.15, 2.153... |
| Probability function | PMF |# PDF |

#### Simple Rule

> **Counting → Discrete**

> **Measuring → Continuous**

### Python Implementation




we will generate a simple discrete dataset representing the number of customers arriving at a store.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

customers = np.random.randint(0, 11, size=100)

print(customers[:20])
```

The number of customers is a discrete variable because it can only take countable values such as:

$$
0,1,2,3,\ldots
$$

A bar chart is suitable for visualizing the frequency of each possible value.

```python
values, counts = np.unique(customers, return_counts=True)

plt.bar(values, counts)

plt.xlabel("Number of Customers")
plt.ylabel("Frequency")
plt.title("Discrete Random Variable")

plt.show()
```


Now consider the waiting time of customers.

Waiting time could be:

$$
2.1,\ 2.15,\ 2.153,\ 2.1537,\ldots
$$

Since waiting time can take many possible values within a range, it is treated as a continuous random variable.

We can simulate waiting times using NumPy.

```python
waiting_time = np.random.uniform(0, 10, size=1000)

print(waiting_time[:10])
```

A histogram is useful for visualizing the distribution of a continuous variable.

Unlike a discrete variable, we are interested in how the values are distributed across ranges or intervals.

```python
plt.hist(waiting_time, bins=20)

plt.xlabel("Waiting Time")
plt.ylabel("Frequency")
plt.title("Continuous Random Variable")

plt.show()
```
