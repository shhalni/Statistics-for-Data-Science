## Cumulative Distribution Function

### 1. Introduction

CDF stands for **Cumulative Distribution Function**.

A CDF tells us the **cumulative probability up to a particular value**.

In simple words:

> **CDF tells us how much probability has been accumulated up to a given value.**

The key idea is:

**CDF = P(X ≤ x)**

This means the probability that the random variable `X` is **less than or equal to `x`**.

"Cumulative" means that we keep **adding probabilities as we move from smaller values to larger values**.


### 2. Intuition


Consider a fair die.

The possible values are:

1, 2, 3, 4, 5, 6

Each value has probability `1/6`.

If we ask:

**What is the probability of getting 3 or less?**

We include:

- P(X = 1)
- P(X = 2)
- P(X = 3)

So:

P(X ≤ 3) = 1/6 + 1/6 + 1/6 = 3/6 = 0.5

Therefore, the CDF at `3` is `0.5`.

So CDF is basically:

> **Start from the beginning and keep adding probability until you reach the value you're interested in.**

### 3. Formula

For a random variable `X`, the CDF is:

$$
F(x) = P(X \leq x)
$$

For a discrete random variable:

$$
F(x) = \sum_{t \leq x} P(X=t)
$$

The CDF always represents **cumulative probability**.

Important properties:

- CDF is always between `0` and `1`.
- CDF never decreases.
- As we move toward the largest possible values, CDF approaches `1`.

### 4. Python Implementation




```python
import numpy as np
import pandas as pd

values = np.array([1, 2, 3, 4, 5, 6])
probabilities = np.array([1/6] * 6)

cdf = np.cumsum(probabilities)

df = pd.DataFrame({
    "Value": values,
    "Probability": probabilities,
    "CDF": cdf
})

df
```

### 5. Understanding the output

For a fair die:

| X | Probability | CDF |
|---|---:|---:|
| 1 | 1/6 | 1/6 |
| 2 | 1/6 | 2/6 |
| 3 | 1/6 | 3/6 |
| 4 | 1/6 | 4/6 |
| 5 | 1/6 | 5/6 |
| 6 | 1/6 | 1 |

Notice that the CDF keeps increasing because we are continuously **accumulating probability**.

For example:

**CDF(3) = P(X ≤ 3) = 3/6 = 0.5**

**CDF(5) = P(X ≤ 5) = 5/6 ≈ 0.833**

So CDF answers questions such as:

- What is the probability of getting 3 or less?
- What is the probability of getting 5 or less?
- What percentage of observations are below a particular value?

### 6. Plot the CDF

```python
import matplotlib.pyplot as plt

plt.step(values, cdf, where="post")

plt.xlabel("Die Value")
plt.ylabel("Cumulative Probability")
plt.title("CDF of a Fair Die")
plt.ylim(0, 1.1)
plt.grid()

plt.show()
```

### 7. CDF for continuos variables


CDF is not limited to discrete variables.

It can also be used with continuous variables such as:

- Height
- Weight
- Temperature
- Age
- Time

For a continuous variable, the CDF tells us:

> **The probability that the value is less than or equal to a particular value.**

For example:

If `X` represents a person's height:

$$
F(170) = P(X \leq 170)
$$

If the result is `0.40`, it means:

> There is a 40% probability that a randomly selected person's height is 170 cm or less.

So CDF is useful for answering **"how much has accumulated up to this value?"**

### 8. PMF vs PDF vs CDF


####  PMF

Used for **discrete variables**.

Tells us the probability of an **individual value**.

Example:

$$
P(X=4)
$$

####  PDF

Used for **continuous variables**.

Tells us the **density** of probability across values.

The **area under the curve over a range** gives probability.

Example:

$$
P(165 < X < 175)
$$

####  CDF

Can be used for both **discrete and continuous variables**.

Tells us the **cumulative probability up to a value**.

Example:

$$
P(X \leq 170)
$$

####  Mental Model

**PMF → probability at individual discrete values**

**PDF → probability density across continuous values**

**CDF → accumulated probability up to a value**n the next section.

### 9. Business Applications

CDF is useful when we want to understand what percentage of observations fall below a particular threshold.

Examples:

- Salary → percentage of employees earning below ₹50,000
- Delivery time → probability that an order arrives within 2 days
- Customer spending → percentage of customers spending less than ₹1,000
- Loan amount → percentage of customers borrowing below a certain amount
- Response time → probability that a system responds within 2 seconds

CDF is especially useful for **threshold-based decisions**.king.ty-weighted average outcomes.

### 10. ML Connection

CDF is useful in data science and machine learning for:

- Understanding the distribution of features
- Finding percentiles and thresholds
- Comparing observations with a population
- Understanding how much data lies below a particular value
- Detecting unusual or extreme observations

For example, a CDF can help answer:

> "What percentage of training data has a feature value below this threshold?"
