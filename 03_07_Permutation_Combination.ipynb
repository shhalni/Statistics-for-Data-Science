## Permutation and Combination

### 1. Introduction

Permutation and Combination are **counting techniques** used to determine how many possible arrangements or selections can be made from a given set of objects.

The main difference is:

- **Permutation → Order matters**
- **Combination → Order does not matter**

These concepts are useful in probability because they allow us to count possible outcomes without manually listing every possibility.

They are used in areas such as probability, experiment design, sampling, scheduling, security, and machine learning.


## 1.Permutation


### 1. Introduction

A **Permutation** is an arrangement of objects where the **order matters**.

For example, suppose we have three people:

A, B, C

If we select two people for two different positions:

- AB
- BA

are considered **different permutations** because the order is different.

Therefore, permutations are used when changing the order creates a different outcome.


### 2. Mathematical Perspective


The number of ways to arrange \(r\) objects from \(n\) objects is:

$$
{}^nP_r = \frac{n!}{(n-r)!}
$$

where:

- \(n\) = Total number of objects
- \(r\) = Number of objects being arranged
- \(!\) = Factorial

#### Factorial

Factorial is the product of all positive integers up to a given number.

$$
n! = n \times (n-1) \times (n-2) \times \cdots \times 1
$$

For example:

$$
5! = 5 \times 4 \times 3 \times 2 \times 1 = 120
$$

### 🧠 Intuition


Imagine that 5 students are competing for three different positions:

- President
- Vice President
- Secretary

If student A becomes President and student B becomes Vice President, this is different from B becoming President and A becoming Vice President.

The selected students are the same, but their **positions are different**.

Therefore, the order matters and we use **Permutation**.

#### Think:

**Arrange → Position → Rank → Order matters → Permutation**

### 3.Example

Suppose there are **5 students** and we need to assign them to three different positions:

- President
- Vice President
- Secretary

Here:

$$
n=5
$$

$$
r=3
$$

Using the permutation formula:

$$
{}^5P_3 = \frac{5!}{(5-3)!}
$$

$$
= \frac{5!}{2!}
$$

$$
= \frac{120}{2}
$$

$$
=60
$$

Therefore, there are **60 different ways** to assign the three positions.our new sample space.

## 2. Combination

### 1. Introduction

A **Combination** is a selection of objects where the **order does not matter**.

If changing the order does not create a new outcome, we use combination.

#### Example

Suppose we have three people:

$$
A,\ B,\ C
$$

We want to select two people for a team.

The following selections are considered the same:

$$
AB = BA
$$

Both represent the same team containing A and B.

Therefore:

> **Combination → Order Does Not Matter**

### 2. Intuition

Imagine that 5 students need to be selected for a project team.

Suppose we select:

$$
A,\ B,\ C
$$

The following are all the same team:

$$
A,B,C
$$

$$
B,A,C
$$

$$
C,B,A
$$

The order in which the students are listed does not matter.

We only care about **who was selected**.

Therefore, we use **Combination**.

#### Think:

**Select → Choose → Team → Order does not matter → Combination** does not matter → Combination**

### 3. Mathematical Formula

The number of ways to select \(r\) objects from \(n\) objects is:

$$
{}^nC_r = \frac{n!}{r!(n-r)!}
$$

where:

- \(n\) = Total number of objects
- \(r\) = Number of objects being selected
- \(!\) = Factorial

For example:

$$
{}^3C_2 = \frac{3!}{2!(3-2)!}
$$

$$
= \frac{3 \times 2 \times 1}{(2 \times 1)(1)}
$$

$$
= \frac{6}{2} = 3
$$

### 4. Example

Suppose there are **5 students** and we need to select **3 students for a project team**.

Here:

$$
n=5
$$

$$
r=3
$$

Using the combination formula:

$$
{}^5C_3 = \frac{5!}{3!(5-3)!}
$$

$$
= \frac{5!}{3!2!}
$$

$$
= \frac{120}{6\times2}
$$

$$
=10
$$

Therefore, there are **10 different teams** that can be formed.

## 2. Permutation vs Combination

| Feature | Permutation | Combination |
|---|---|---|
| Meaning | Arrangement | Selection |
| Does order matter? | Yes | No |
| Formula | $$
{}^nP_r = \frac{n!}{(n-r)!}
$$| $$
{}^nC_r = \frac{n!}{r!(n-r)!}
$$ |
| Example | Assigning positions | Selecting a team |
| Key idea | Arrange | Choose |

#### Simple Rule

Ask yourself:

> **Does changing the order create a different outcome?**

If **Yes** → Permutation

If **No** → Combination

## 3.Relationship b/w Permutation and Combination

Permutation and Combination are closely related.

First, we can select \(r\) objects using a combination:

$$
{}^nC_r
$$

Then, we can arrange those \(r\) selected objects in:

$$
r!
$$

different ways.

Therefore:

$$
{}^nP_r = {}^nC_r \times r!
$$

This explains why permutation produces more arrangements than combination for the same \(n\) and \(r\).

#### Example

For 5 objects and selecting 3:

$$
{}^5C_3 = 10
$$

Each group of 3 can be arranged in:

$$
3! = 6
$$

ways.

Therefore:

$$
{}^5P_3 = 10\times6 = 60
$$

## 4. Python Implementation




Python provides built-in functions for calculating permutations and combinations through the `math` module.

- `math.perm(n, r)` → Permutation
- `math.comb(n, r)` → Combination

```python
import math

n = 5
r = 3

permutation = math.perm(n, r)
combination = math.comb(n, r)

print("Permutation:", permutation)
print("Combination:", combination)
```

## Generating the Actual Arrangements and Selections

The formulas tell us **how many** possibilities exist.

Python can also generate the actual permutations and combinations using the `itertools` module.

For example, consider:

$$
\{A,B,C\}
$$

and select 2 objects.

```python
from itertools import permutations, combinations

items = ["A", "B", "C"]

permutation_result = list(permutations(items, 2))
combination_result = list(combinations(items, 2))

print("Permutations:")
print(permutation_result)

print("\nCombinations:")
print(combination_result)
```

### Interpretation

For permutations:

$$
(A,B) \neq (B,A)
$$

Therefore, both are included.

For combinations:

$$
(A,B) = (B,A)
$$

Therefore, only one of them is included.

This demonstrates the fundamental difference:

> **Permutation → Order matters**

> **Combination → Order does not matter**

##  Connection to Probability


Permutations and combinations are often used to count possible outcomes when calculating probabilities.

#### Example

Suppose we select 3 students from 10 students to form a team.

Since the order of the students does not matter, we use combination:

$$
{}^{10}C_3
$$

However, if those 3 students are assigned to three different positions, such as:

- President
- Vice President
- Secretary

then order matters, so we use permutation:

$$
{}^{10}P_3
$$

Therefore, the type of counting technique depends on whether the order changes the outcome.
