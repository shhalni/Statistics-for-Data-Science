# Hypothesis Testing 



## 1. What is hypothesis testing?

Hypothesis testing is a statistical method used to evaluate a claim about a population using sample evidence.

A typical workflow:
1. Define the research question and population parameter.
2. State the null and alternative hypotheses.
3. Choose a significance level (`alpha`).
4. Select an appropriate statistical test.
5. Calculate the test statistic and p-value.
6. Make a decision about the null hypothesis.
7. Interpret the result in context.


## 2. Null and alternative hypotheses

- **Null hypothesis (H₀):** The default claim, often representing no difference, no effect, or a specified value.
- **Alternative hypothesis (H₁ or Hₐ):** The claim for which we seek evidence.

Example: Is the population mean mathematics score different from 70?

$$
H_0: \mu = 70
$$

$$
H_1: \mu \ne 70
$$

| Test direction | Alternative hypothesis | Meaning |
|---|---|---|
| Two-tailed | $\mu \ne 70$ | Mean is different |
| Left-tailed | $\mu < 70$ | Mean is lower |
| Right-tailed | $\mu > 70$ | Mean is higher |

Choose the direction from the research question before examining the result.


## 3. Significance level (alpha)

The significance level, denoted by $\alpha$, is the threshold used to determine whether the observed evidence is sufficiently inconsistent with the null hypothesis ($H_0$).

A commonly used significance level is:

$$
\alpha = 0.05
$$

If the null hypothesis ($H_0$) is true, $\alpha$ represents the long-run probability of committing a Type I error when using the test procedure.

**Important:** The significance level ($\alpha$) must be chosen before conducting the hypothesis test.

Alpha is **not** the probability that H₀ is true.


## 4. P-value

The **p-value** is the probability, assuming H₀ is true, of obtaining a test statistic at least as extreme as the observed one, in the direction(s) specified by H₁.

Decision rule:
- If `p-value <= alpha`: reject H₀.
- If `p-value > alpha`: fail to reject H₀.

A p-value is not the probability that H₀ is true. “Fail to reject H₀” does not prove H₀ is true.


## 5. Type I error, Type II error, and power

| Concept | Meaning |
|---|---|
| Type I error | Rejecting a true H₀ (false positive) |
| Type II error | Failing to reject a false H₀ (false negative) |
| Alpha (α) | Probability of Type I error under H₀ |
| Beta (β) | Probability of Type II error under a specified alternative |
| Power | Probability of correctly rejecting a false H₀ |

$$
\text{Power} = 1 - \beta
$$

Power is influenced by sample size, effect size, variability, and significance level.


## 6. Test statistic and critical value

A test statistic measures how far the observed sample result is from what H₀ predicts, expressed in units of estimated standard error.

$$
\text{Test statistic} =
\frac{\text{Observed estimate} - \text{Value under }H_0}
{\text{Standard error}}
$$

The **critical-value approach** compares the test statistic with a cutoff. The **p-value approach** compares the p-value with alpha. When applied correctly, both approaches lead to the same decision.


## 7. Degrees of freedom

Degrees of freedom (df) describe how many independent pieces of information are available to estimate variability after accounting for constraints.

For a one-sample t-test:

\[
df = n - 1
\]

Degrees of freedom help determine the appropriate t-distribution and its critical values.


## 8. Choosing a statistical test

| Research question / data | Common test |
|---|---|
| One sample mean vs a hypothesised value | One-sample t-test |
| Means of two independent groups | Independent two-sample t-test |
| Before-and-after measurements on the same subjects | Paired t-test |
| One population proportion vs a hypothesised proportion | One-proportion z-test |
| Proportions from two independent groups | Two-proportion z-test |
| Association between two categorical variables | Chi-square test of independence |
| Means across three or more independent groups | One-way ANOVA |
| Linear association between two numeric variables | Pearson correlation test |

Test choice depends on the research question, variable types, study design, and assumptions.


## 9. Assumptions and checks

**T-tests**
- Observations should be independent based on the study design.
- The outcome should be numeric.
- For small samples, the relevant population or paired differences should be reasonably close to normal, without severe outliers.
- Welch's independent t-test does not require equal group variances.

**Proportion z-tests**
- Outcomes are binary (success/failure).
- Observations should be independent.
- Expected successes and failures should be sufficiently large for the normal approximation.

**Chi-square tests**
- Variables are categorical.
- Observations should be independent.
- Expected cell counts should be sufficiently large for the usual approximation.

**ANOVA**
- Independent observations.
- Numeric outcome.
- Residuals within groups should be reasonably normal.
- Similar group variances for classical one-way ANOVA.

A test function returning a p-value does not verify its assumptions.

## 11. One-sample t-test

**Question:** Is the mean mathematics score different from 70?

$$
H_0: \mu = 70
\qquad
H_1: \mu \ne 70
$$

Use a one-sample t-test to compare one sample mean with a hypothesised value when the population standard deviation is unknown.


```python
scores = df["math score"].dropna()
hypothesised_mean = 70
alpha = 0.05

t_statistic, p_value = stats.ttest_1samp(
    scores,
    popmean=hypothesised_mean,
    alternative="two-sided"
)

print("Sample size:", len(scores))
print("Sample mean:", scores.mean())
print("Sample standard deviation:", scores.std(ddof=1))
print("Degrees of freedom:", len(scores) - 1)
print("t-statistic:", t_statistic)
print("p-value:", p_value)

if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

## 12. Independent two-sample t-test (Welch's test)

**Question:** Is the average mathematics score different between female and male students?

The groups contain different students, so this is an independent-groups comparison. Welch's t-test (`equal_var=False`) does not assume equal population variances.


```python
female_scores = df.loc[df["gender"] == "female", "math score"].dropna()
male_scores = df.loc[df["gender"] == "male", "math score"].dropna()

t_statistic, p_value = stats.ttest_ind(
    female_scores,
    male_scores,
    equal_var=False,
    alternative="two-sided"
)

print("Female mean:", female_scores.mean())
print("Male mean:", male_scores.mean())
print("t-statistic:", t_statistic)
print("p-value:", p_value)
```

## 13. Paired t-test

**Use when:** Two measurements come from the same subjects or genuinely matched pairs.

**Example:** The same students take a test before and after a preparation programme.

$$
H_0: \mu_d = 0
\qquad
H_1: \mu_d \ne 0
$$

Here, \(d\) is the within-person difference (after − before).

The Student Performance dataset does not contain before-and-after measurements for the same students. The following values are illustrative only.


```python
# Illustrative paired data only — not from StudentsPerformance.csv
before_scores = np.array([62, 70, 68, 75, 59, 81, 73, 66, 71, 64])
after_scores  = np.array([68, 74, 72, 79, 65, 84, 78, 70, 76, 69])

assert len(before_scores) == len(after_scores)

differences = after_scores - before_scores

t_statistic, p_value = stats.ttest_rel(
    after_scores,
    before_scores,
    alternative="two-sided"
)

print("Mean before:", before_scores.mean())
print("Mean after:", after_scores.mean())
print("Mean paired difference:", differences.mean())
print("t-statistic:", t_statistic)
print("p-value:", p_value)
```

## 14. One-proportion z-test

**Question:** Is the proportion of students who completed the test preparation course different from 40%?

$$
H_0: p = 0.40
\qquad
H_1: p \ne 0.40
$$

- `count`: number of successes.
- `nobs`: total number of non-missing observations.
- `value`: hypothesised population proportion under H₀.

This implementation uses `statsmodels.stats.proportion.proportions_ztest`.


```python
count = (df["test preparation course"] == "completed").sum()
nobs = df["test preparation course"].notna().sum()

p0 = 0.40
alpha = 0.05

z_statistic, p_value = proportions_ztest(
    count=count,
    nobs=nobs,
    value=p0,
    alternative="two-sided"
)

print("Number of successes:", count)
print("Number of observations:", nobs)
print("Sample proportion:", count / nobs)
print("Hypothesised proportion:", p0)
print("Z-statistic:", z_statistic)
print("P-value:", p_value)

if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

## 15. Two-proportion z-test

**Question:** Is the proportion completing test preparation different between female and male students?

$$
H_0: p_1 = p_2
\qquad
H_1: p_1 \ne p_2
$$

The two groups must be independent. Each group needs a success count and a total number of non-missing observations. The usual two-proportion z-test pools the proportion under H₀.


```python
female = df[df["gender"] == "female"]
male = df[df["gender"] == "male"]

female_count = (female["test preparation course"] == "completed").sum()
male_count = (male["test preparation course"] == "completed").sum()

female_n = female["test preparation course"].notna().sum()
male_n = male["test preparation course"].notna().sum()

counts = np.array([female_count, male_count])
sample_sizes = np.array([female_n, male_n])

z_statistic, p_value = proportions_ztest(
    count=counts,
    nobs=sample_sizes,
    alternative="two-sided"
)

print("Female completion proportion:", female_count / female_n)
print("Male completion proportion:", male_count / male_n)
print("Z-statistic:", z_statistic)
print("P-value:", p_value)

if p_value <= 0.05:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

## 16. Chi-square goodness-of-fit test

**Use when:** You have one categorical variable and want to compare its observed category counts with expected counts from a specified distribution.

**Example question:** Are the four observed category counts consistent with an equal distribution across the four categories?

- \(H_0\): The population follows the specified distribution (here, each category has probability \(0.25\)).
- \(H_1\): The population distribution differs from the specified distribution.

The test compares observed counts (\(O\)) with expected counts (\(E\)):

$$
\chi^2 = \sum \frac{(O_i-E_i)^2}{E_i}
$$

For this example, expected counts are equal because the null hypothesis specifies equal proportions. In other applications, expected proportions can be unequal and should be specified from the research question or a justified model.

```python
# Example observed counts for one categorical variable
# Replace these example counts with your own observed category frequencies.
observed = np.array([40, 30, 20, 10])

# Under H0, all four categories are equally likely
expected = np.repeat(observed.sum() / len(observed), len(observed))

chi2_statistic, p_value = stats.chisquare(
    f_obs=observed,
    f_exp=expected
)

degrees_freedom = len(observed) - 1
alpha = 0.05

print("Observed counts:", observed)
print("Expected counts:", expected)
print("Chi-square statistic:", chi2_statistic)
print("Degrees of freedom:", degrees_freedom)
print("P-value:", p_value)

if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

### Goodness-of-fit vs independence

| Chi-square goodness-of-fit | Chi-square test of independence |
|---|---|
| One categorical variable | Two categorical variables |
| Compares observed counts with specified expected proportions | Tests whether two categorical variables are associated |
| Example: Do category counts follow an equal distribution? | Example: Is gender associated with test-preparation completion? |
| Python: `stats.chisquare()` | Python: `stats.chi2_contingency()` |

**Assumptions and checks:** observations should be independent, categories should be mutually exclusive, and expected counts should generally be sufficiently large for the chi-square approximation. If expected counts are too small, consider an appropriate exact or simulation-based method.


## 17. Chi-square test of independence

**Question:** Is gender associated with completion of the test preparation course?

Both variables are categorical. The test evaluates whether they are statistically associated; it does not establish causation.


```python
contingency_table = pd.crosstab(
    df["gender"],
    df["test preparation course"]
)

chi2_statistic, p_value, degrees_freedom, expected = stats.chi2_contingency(
    contingency_table
)

print("Observed counts:")
display(contingency_table)

print("Expected counts:")
display(pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
))

print("Chi-square statistic:", chi2_statistic)
print("Degrees of freedom:", degrees_freedom)
print("P-value:", p_value)
```

## 18. One-way ANOVA

**Question:** Does the average mathematics score differ across parental education groups?

- H₀: All group population means are equal.
- H₁: At least one group mean differs.

A significant ANOVA result does not identify which specific groups differ. A suitable post-hoc analysis is needed for that.


```python
groups = [
    group["math score"].dropna().to_numpy()
    for _, group in df.groupby("parental level of education")
]

f_statistic, p_value = stats.f_oneway(*groups)

print("F-statistic:", f_statistic)
print("p-value:", p_value)
```

## 19. Pearson correlation significance test

**Question:** Is there evidence of a linear association between math and reading scores?

Pearson's correlation test evaluates whether the population Pearson correlation is zero, under its assumptions. Correlation does not establish causation.


```python
paired_scores = df[["math score", "reading score"]].dropna()

r_statistic, p_value = stats.pearsonr(
    paired_scores["math score"],
    paired_scores["reading score"]
)

print("Pearson correlation:", r_statistic)
print("p-value:", p_value)
```

## 20. Effect size

A p-value indicates how compatible the observed data are with H₀; it does not tell us how large or practically important an effect is.

Examples of effect sizes:
- Difference between two means.
- Cohen's d for a standardised mean difference.
- Difference between two proportions.
- Cramér's V for association in a contingency table.
- Correlation coefficient `r` for linear association.

Report effect size alongside statistical significance when appropriate.


## 21. Confidence intervals and hypothesis testing

Confidence intervals provide a range of plausible values for a population parameter. They complement hypothesis tests by showing the estimated effect and its precision.

For a two-sided test at significance level 0.05, a corresponding 95% confidence interval and test of the same parameter are consistent: if the null value lies outside the interval, the test rejects that null value at the 5% level, subject to matching methods and assumptions.


## 22. Multiple testing and post-hoc tests

When many hypotheses are tested, the chance of obtaining at least one false positive can increase.

- **Bonferroni correction:** adjusts the significance threshold for multiple comparisons.
- **False discovery rate (FDR):** controls the expected proportion of false discoveries among rejected hypotheses.
- **Post-hoc tests:** after a significant ANOVA, suitable pairwise comparisons can identify which groups differ.

Choose and report the correction method rather than interpreting many unadjusted p-values as if only one test had been performed.


## 23. Non-parametric alternatives

Non-parametric tests can be useful when parametric assumptions are unsuitable, depending on the study design and question.

| Situation | Possible non-parametric test |
|---|---|
| Two independent groups | Mann–Whitney U |
| Paired measurements | Wilcoxon signed-rank |
| Three or more independent groups | Kruskal–Wallis |

These tests are not automatic replacements for t-tests or ANOVA in every situation. Their interpretation and assumptions differ.


## 24. Reporting results responsibly

A useful statistical conclusion includes:
1. Research question and hypotheses.
2. Test used.
3. Test statistic and degrees of freedom, where relevant.
4. P-value.
5. Decision at the preselected alpha level.
6. Conclusion in context.
7. Relevant assumptions and limitations.
8. Effect size or confidence interval where appropriate.

Avoid saying “accept H₀.” Prefer **“fail to reject H₀.”**

Statistical significance is not the same as practical importance.
