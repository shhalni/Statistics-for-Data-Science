## 3. Sample Space

### 1. Introduction

A **Sample Space** is the **set of all possible outcomes** of a random experiment.

It is usually denoted by the symbol **S**.

Every outcome produced by a random experiment belongs to the sample space.

Without knowing the sample space, we cannot calculate probabilities because probability is always based on the possible outcomes of an experiment.

---



### 2. Intuition


Imagine rolling a fair six-sided die.

Before rolling, you don't know which number will appear.

However, you know that the result must be one of these six numbers:

$$
S=\{1,2,3,4,5,6\}
$$

If the die lands on **4**, then

- Sample Space = {1,2,3,4,5,6}
- Outcome = 4

The sample space contains **every possible result**, while an outcome is **one result** obtained from the experiment.

---

#### 🌍 Examples

| Random Experiment | Sample Space |
|-------------------|--------------|
| Toss a Coin | {Heads, Tails} |
| Roll a Die | {1,2,3,4,5,6} |
| Draw a Card | All 52 playing cards |
| Traffic Signal | {Red, Yellow, Green} |
| Weather Tomorrow | {Rain, No Rain} |
| Customer Purchase | {Purchase, No Purchase} |

---


### 3. Mathematical Perspective


For a coin toss,

$$
S=\{H,T\}
$$

where

- H = Heads
- T = Tails

---

For rolling a die,

$$
S=\{1,2,3,4,5,6\}
$$

---

For a traffic signal,

$$
S=\{\text{Red},\text{Yellow},\text{Green}\}
$$

---

#### 📝 Key Points

- Sample space is the collection of all possible outcomes.
- It is represented by **S**.
- Every outcome belongs to the sample space.
- Probability is defined using the sample space.

### 4. Python Implementation




#### Coin Toss Simulation

```python
coin_sample_space = ["Heads", "Tails"]

print("Sample Space of Coin Toss")
print(coin_sample_space)
```

#### Dice Roll Simulation

```python
dice_sample_space = list(range(1,7))

print("Sample Space of Rolling a Die")
print(dice_sample_space)
```


#### Generate Sample Space

```python
colors = ["Red", "Blue", "Green"]

print("Sample Space")

for color in colors:
    print(color)
```

### 5. Visualization



```python
import matplotlib.pyplot as plt

sample_space = ["Heads", "Tails"]
frequency = [1,1]

plt.figure(figsize=(5,4))
plt.bar(sample_space, frequency)

plt.title("Sample Space of a Coin Toss")
plt.xlabel("Possible Outcomes")
plt.ylabel("Presence")

plt.show()
```

#### Interpretation

- Each bar represents one possible outcome.
- A sample space lists all possible outcomes, not how often they occur.
- Every future outcome of the experiment must belong to this sample space.

### 8. Machine Learning Connection



Machine Learning models are designed to predict one outcome from a predefined sample space.

Examples

| Machine Learning Problem | Sample Space |
|--------------------------|--------------|
| Spam Detection | {Spam, Not Spam} |
| Disease Prediction | {Disease, Healthy} |
| Loan Approval | {Approve, Reject} |
| Customer Churn | {Churn, No Churn} |
| Image Classification | {Cat, Dog} |

Before training a model, we define all possible prediction classes. These classes together form the sample space.

For example, in the Titanic dataset,

Target Variable:

**Survived**

Sample Space:

$$
S=\{0,1\}
$$

where

- 0 → Did Not Survive
- 1 → Survived

Machine learning algorithms learn from historical data to predict which outcome from this sample space is most likely for a new observation.
