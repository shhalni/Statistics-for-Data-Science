## 3. Outcome

### 1. Introduction

An **Outcome** is the **result obtained after performing a random experiment**.

Every time a random experiment is conducted, **exactly one outcome** occurs from the set of all possible outcomes (called the **sample space**).

For example, when rolling a fair six-sided die, the possible outcomes are:

$$
S=\{1,2,3,4,5,6\}
$$

If the die lands on **4**, then **4** is the outcome of that experiment.

Similarly,

- Tossing a coin → Outcome: **Heads** or **Tails**
- Drawing a card → Outcome: **Ace of Spades**
- Selecting a passenger from the Titanic dataset → Outcome: **Survived** or **Did Not Survive**

An outcome is the **smallest unit** in probability theory because probabilities are assigned to outcomes or groups of outcomes.

---


### 2. Intuition


Imagine tossing a coin.

Before tossing, there are two possible outcomes:

- Heads
- Tails

Suppose the coin lands on **Heads**.

Then,

- **Random Experiment:** Tossing a Coin
- **Outcome:** Heads

Only **one outcome** occurs in a single trial.

Similarly,

If you roll a die once and obtain **5**, then **5** is the outcome.

If you roll it again, the outcome may be different.

This uncertainty is what probability helps us measure.

### 3. Mathematical Perspective


For a coin toss,

$$
S=\{H,T\}
$$

If the coin lands on **Heads**,

$$
\text{Outcome}=H
$$

---

For a die,

$$
S=\{1,2,3,4,5,6\}
$$

If the die lands on **3**,

$$
\text{Outcome}=3
$$

---

#### 📝 Key Points

- Every random experiment produces exactly one outcome.
- An outcome is one element of the sample space.
- Outcomes are used to calculate probabilities.
- Probability assigns a likelihood to different outcomes.

### 4. Python Implementation




#### Coin Toss Simulation

```python
import random

outcome = random.choice(["Heads", "Tails"])

print("Random Experiment : Toss a Coin")
print("Outcome :", outcome)
```

#### Dice Roll Simulation

```python
import random

outcome = random.randint(1,6)

print("Random Experiment : Roll a Die")
print("Outcome :", outcome)
```


#### Multiple Outcomes

```python
import random

coin = ["Heads", "Tails"]

print("10 Coin Tosses\n")

for i in range(10):
    print(f"Trial {i+1}: {random.choice(coin)}")
```

### 5. Visualization



```python
import random
import matplotlib.pyplot as plt
from collections import Counter

# Simulate 500 dice rolls
rolls = [random.randint(1,6) for _ in range(500)]

counts = Counter(rolls)

plt.figure(figsize=(7,4))

plt.bar(counts.keys(), counts.values(), edgecolor="black")

plt.title("Frequency of Outcomes (500 Dice Rolls)")
plt.xlabel("Outcome")
plt.ylabel("Frequency")

plt.xticks(range(1,7))

plt.grid(axis='y', alpha=0.3)

plt.show()
```

#### Interpretation

- Each bar represents one possible outcome of the random experiment.
- Since the die is fair, each outcome occurs with approximately the same frequency.
- As the number of trials increases, the observed frequencies approach the theoretical probabilities.

### 8. Machine Learning Connection



Machine Learning models are designed to predict one outcome from a predefined sample space.

Examples

| Machine Learning Problem | Sample Space |
|--------------------------|--------------|
| Spam Detection | {Spam, Not Spam} |
| Disease Prediction | {Disease, Healthy} |
| Loan Approval | {Approve, Reject} |
| Customer Churn | {Churn, No Churn} |
| Image Classification | {Cat, Dog} |

Before training a model, we define all possible prediction classes. These classes together form the sample space.

For example, in the Titanic dataset,

Target Variable:

**Survived**

Sample Space:

$$
S=\{0,1\}
$$

where

- 0 → Did Not Survive
- 1 → Survived

Machine learning algorithms learn from historical data to predict which outcome from this sample space is most likely for a new observation.
