## 4. Event

### 1. Introduction

An **Event** is a **set of one or more outcomes** from the sample space that satisfy a particular condition.

In probability, we are usually interested in events rather than individual outcomes.

For example, when rolling a fair six-sided die,

$$
S=\{1,2,3,4,5,6\}
$$

Suppose we are interested in getting an **even number**.

The event is

$$
E=\{2,4,6\}
$$

Here,

- Sample Space contains all possible outcomes.
- Event contains only the outcomes that satisfy the required condition.

An event may consist of:

- A single outcome.
- Multiple outcomes.
- All outcomes.
- No outcomes.

---


### 2. Intuition


Imagine rolling a die.

You are not interested in the exact number.

Instead, someone asks,

> "Did you roll an even number?"

Now you don't care whether it was **2**, **4**, or **6**.

As long as the outcome belongs to

$$
\{2,4,6\}
$$

the event has occurred.

Similarly,

Random Experiment:

Roll a Die

Outcome:

4

Event:

Getting an Even Number

Since 4 belongs to {2,4,6}, the event occurs.

---

### 📝 Key Points

- An event is a subset of the sample space.
- An event contains one or more outcomes.
- Events are used to calculate probabilities.
- If the observed outcome belongs to the event, the event is said to occur.

### 4. Python Implementation




#### Even Number Event

```python
sample_space = [1,2,3,4,5,6]

event = [x for x in sample_space if x % 2 == 0]

print("Sample Space :", sample_space)
print("Event (Even Numbers):", event)
```

#### Odd Number Event

```python
sample_space = [1,2,3,4,5,6]

event = [x for x in sample_space if x % 2 != 0]

print(event)
```

#### Numbers > 4

```python
sample_space = [1,2,3,4,5,6]

event = [x for x in sample_space if x > 4]

print(event)
```

#### Coin Toss

```python
sample_space = ["Heads","Tails"]

event = ["Heads"]

print("Sample Space:", sample_space)
print("Event:", event)
```

## Types of Event

An **event** is a collection of one or more outcomes from the sample space. Based on their characteristics, events can be classified into different types.

Understanding these event types is essential because they form the foundation for probability rules such as the Addition Rule, Multiplication Rule, Conditional Probability, and Bayes' Theorem.

The main types of events are:

1. Simple Event
2. Compound Event
3. Certain Event
4. Impossible Event
5. Complementary Event
6. Mutually Exclusive Events
7. Exhaustive Events

### 1. Simple Event

#### 📖 Concept

A **Simple Event** is an event that contains **exactly one outcome** from the sample space.

Example:

Rolling a die

$$
S=\{1,2,3,4,5,6\}
$$

Event:

Getting a **4**

$$
E=\{4\}
$$

Since the event contains only one outcome, it is called a simple event.

#### 🧠 Intuition

Think of a simple event as asking for one specific result.

Example:

> Roll a die and get exactly **6**.

Only one outcome satisfies the condition.

#### 🌍 Examples

- Toss a coin → Getting Heads
- Roll a die → Getting 5
- Draw a card → Ace of Spades

```python
sample_space = [1,2,3,4,5,6]

simple_event = [4]

print(simple_event)
```

### 2. Compound Event

#### 📖 Concept

A **Compound Event** contains **more than one outcome**.

Example

Rolling a die

Event:

Getting an even number

$$
E=\{2,4,6\}
$$

The event contains three outcomes.

Hence, it is called a compound event.

#### 🧠 Intuition

Instead of asking for one number,

we ask for a group of numbers satisfying a condition.

Example:

"Roll an even number."

Any of

2

4

6

will satisfy the event.

```python
sample_space = [1,2,3,4,5,6]

compound_event = [x for x in sample_space if x%2==0]

print(compound_event)
```

### 3. Certain Event

#### 📖 Concept

A **Certain Event** is an event that is guaranteed to occur.

Its probability is

$$
P(E)=1
$$

Example

Rolling a die.

Event:

Getting a number less than 7.

Every possible outcome satisfies this condition.

$$
E=\{1,2,3,4,5,6\}
$$

```python
sample_space = [1,2,3,4,5,6]

certain_event = [x for x in sample_space if x<7]

print(certain_event)
```

### 4. Impossible Event

#### 📖 Concept

An **Impossible Event** is an event that can never occur.

Its probability is

$$
P(E)=0
$$

Example

Rolling a die.

Event:

Getting an 8.

Since 8 is not in the sample space,

the event is impossible.

$$
E=\{\}
$$

```python
sample_space = [1,2,3,4,5,6]

impossible_event = [x for x in sample_space if x==8]

print(impossible_event)
```

### 5. Complementary Event

#### 📖 Concept

The **Complementary Event** consists of all outcomes that are **not part of the given event**.

If

$$
E=\{2,4,6\}
$$

then

$$
E'=\{1,3,5\}
$$

The probability of the complement is

$$
P(E')=1-P(E)
$$

```python
sample_space = [1,2,3,4,5,6]

event = [2,4,6]

complement = [x for x in sample_space if x not in event]

print(complement)
```

### 6. Mutually Exclusive Event

#### 📖 Concept

Two events are **Mutually Exclusive** if they **cannot occur at the same time**.

Example

Rolling a die

A = Getting an even number

$$
A=\{2,4,6\}
$$

B = Getting an odd number

$$
B=\{1,3,5\}
$$

No outcome belongs to both events.

Therefore,

A and B are mutually exclusive.

```python
A = {2,4,6}

B = {1,3,5}

print(A.intersection(B))
```

### 7. Exhaustive Events

#### 📖 Concept

Events are called **Exhaustive** if together they cover the **entire sample space**.

Example

Coin Toss

Sample Space

$$
S=\{H,T\}
$$

Events

A = Heads

B = Tails

Together,

$$
A\cup B=S
$$

Hence,

they are exhaustive events.

```python
sample_space = {"Heads","Tails"}

A = {"Heads"}

B = {"Tails"}

print(A.union(B))
```
