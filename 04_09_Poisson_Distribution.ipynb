## Poisson distribution

### 1. Introduction

The **Poisson Distribution** is used to model the number of times an event occurs within a **fixed interval**.

The interval can be:

- Time → number of calls in 1 hour
- Space → number of defects in 1 meter of fabric
- Area → number of accidents in 1 km of road
- Website traffic → number of requests in 1 minute

The main question is:

> **"What is the probability of getting a certain number of events in a fixed interval?"**

For example:

> A website receives an average of 5 orders per hour. What is the probability of receiving exactly 3 orders in the next hour?


### 2. Intuition


Suppose a restaurant receives an average of **10 customers per hour**.

We can ask:

- What is the probability of getting exactly 5 customers in an hour?
- What is the probability of getting exactly 10 customers?
- What is the probability of getting 15 customers?

We are **counting how many times an event occurs** during a fixed interval.

The event is:

**Customer arrives**

The interval is:

**1 hour**

The count is:

**Number of customers**

This is a Poisson Distribution situation.

### Mental Model

**Poisson = Count events within a fixed interval**


#### The λ (Lambda) Parameter

Poisson Distribution uses a parameter called **λ (lambda)**.

`λ` represents the **average number of events occurring in the given interval**.

For example:

If a call center receives an average of 20 calls per hour:

$$
\lambda = 20
$$

If the interval is one hour, the average is 20 calls per hour.

If we change the interval, λ changes accordingly.

For example:

- 20 calls/hour → λ = 20 for 1 hour
- 10 calls/30 minutes → λ = 10 for 30 minutes
- 5 calls/15 minutes → λ = 5 for 15 minutes

#### Important

For a Poisson Distribution:

**Mean = λ**

**Variance = λ**


### 3. Conditions


Poisson Distribution is appropriate when:

1. We are **counting events**.
2. We consider a **fixed interval** of time, space, area, etc.
3. Events occur independently.
4. The average rate of occurrence is approximately constant.
5. Two events cannot occur at exactly the same instant/location in the idealized model.

Examples:

- Calls per hour
- Customers per minute
- Defects per meter
- Website requests per second

### 4. Formula


The probability of observing exactly `k` events is:

$$
P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}
$$

Where:

- `X` = number of events
- `k` = number of events we want
- `λ` = average number of events in the interval
- `e` ≈ 2.718
- `k!` = factorial of `k`

You do not need to memorize the formula immediately.

The important idea is:

> **λ tells us the average rate, and the distribution tells us the probability of different event counts around that average.**

### 5. Simple Example


Suppose a website receives an average of **5 orders per hour**.

Therefore:

$$
\lambda = 5
$$

We want to find the probability of receiving exactly **3 orders** in one hour.

So:

`k = 3`

We calculate:

$$
P(X=3)=\frac{e^{-5}5^3}{3!}
$$

The probability is approximately:

**0.1404 = 14.04%**

So there is about a **14% probability of receiving exactly 3 orders in one hour**.

### 6. Python Implementation




```python
from scipy.stats import poisson

lam = 5
k = 3

probability = poisson.pmf(k, lam)

print("Probability of exactly 3 orders:", probability)
print("Percentage:", probability * 100)
```

#### Visualization

A Poisson Distribution shows the probability of getting different numbers of events.

For example, if the average number of events is 5, we can see the probability of:

0 events, 1 event, 2 events, 3 events, ... and so on.

The distribution is centered around its average value, λ.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lam = 5

x = np.arange(0, 16)
probabilities = poisson.pmf(x, lam)

plt.bar(x, probabilities)

plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution (λ = 5)")

plt.show()
```

### Bernoulli vs Binomial vs Poisson

#### Bernoulli

**One trial + two outcomes**

Example:

> Will a customer purchase?

Answer: Yes / No

#### Binomial

**Fixed number of trials + count successes**

Example:

> Out of 100 customers, how many will purchase?

#### Poisson

**Count events within a fixed interval**

Example:

> How many customers will arrive in one hour?

#### Mental Model

**Bernoulli → One binary outcome**

**Binomial → Number of successes in n trials**

**Poisson → Number of events in an interval**
