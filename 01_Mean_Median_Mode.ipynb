# Mean, Median , Mode

---

# 1. Introduction

Mean, Median and Mode are the most commonly used measures of **central tendency** in statistics.

They help us identify the **center or typical value** of a dataset.

Understanding central tendency is important because raw data can contain hundreds or thousands of observations, making it difficult to understand the dataset by looking at individual values.

Instead, we can summarize the dataset using a few representative measures.

The three main measures of central tendency are:

- **Mean** – Average value of the observations
- **Median** – Middle value after sorting the observations
- **Mode** – Most frequently occurring value

These measures are widely used in:

- Data Analysis
- Exploratory Data Analysis (EDA)
- Business Analytics
- Data Science
- Machine Learning

# 2. Intuition


Consider the following dataset representing the marks of students:

45, 48, 50, 52, 55

We can see that most students scored around the range of 45–55.

But if we have hundreds of students, simply looking at every value becomes difficult.

This raises an important question:

> **How can we represent the center of the data using a single value?**

This is where Mean, Median and Mode are useful.

### Mean

The mean gives the **average** of all observations.

### Median

The median gives the **middle value** after arranging the observations in ascending order.

### Mode

The mode gives the **most frequently occurring value**.

Each measure describes the center of the data from a different perspective.

# 3. Mathematical Formulae



## 1. Mean

The arithmetic mean is calculated by dividing the sum of all observations by the total number of observations.

### Formula

$$
\bar{x} = \frac{\sum_{i=1}^{n}x_i}{n}
$$

Where:

- $x_i$ = Individual observation
- $n$ = Number of observations
- $\bar{x}$ = Mean

---

## 2. Median

Median is the middle value of an ordered dataset.

### Odd Number of Observations

If $n$ is odd:

$$
Median = x_{\frac{n+1}{2}}
$$

### Even Number of Observations

If $n$ is even:

$$
Median =
\frac{x_{\frac{n}{2}} + x_{\frac{n}{2}+1}}{2}
$$

---

## 3. Mode

Mode is the value that occurs most frequently in a dataset.

A dataset can have:

- One mode → Unimodal
- Two modes → Bimodal
- Multiple modes → Multimodal
- No unique mode

### Interpretation

- The **Mean** represents the average marks.
- The **Median** represents the middle observation.
- The **Mode** represents the most frequently occurring score.
- The relative position of Mean and Median can provide an indication of skewness.

# 9. Business Applications

Mean, Median and Mode are widely used in business analytics.

### Mean

Used for:

- Average sales
- Average revenue
- Average customer spending
- Average order value
- Average delivery time

### Median

Useful for:

- Customer income
- House prices
- Salary analysis
- Delivery times
- Purchase amounts

Especially useful when the data contains outliers.

### Mode

Useful for:

- Most purchased product
- Most common customer category
- Most common shoe size
- Most popular payment method
- Most frequently selected product color

### Example

An e-commerce company wants to understand customer spending.

If a few customers spend extremely large amounts, the mean may be much higher than what a typical customer spends.

Median can provide a better representation of the typical customer.

# 10. Machine Learning Connection

Mean, Median and Mode are important in data preprocessing and exploratory data analysis.

### 1. Missing Value Imputation

Missing numerical values can sometimes be replaced using:

- Mean
- Median

Median is often preferred when the feature contains strong outliers.

### 2. Exploratory Data Analysis

Mean, median and mode help us understand:

- Center
- Distribution
- Skewness
- Outliers

### 3. Feature Engineering

Understanding the central tendency of features helps us decide how to transform and prepare data.

### 4. Categorical Data

Mode can be used to understand the most common category and can sometimes be used for categorical missing-value imputation.

### Important

Mean, median and mode are descriptive statistics. They help us understand and prepare data before applying machine learning models.
