## Discrete probability distribution

### 1. Introduction

A **discrete probability distribution** describes the probabilities of the possible values of a **discrete random variable**.

A discrete random variable has **countable possible values**.

Examples:

- Number of heads in coin tosses
- Number of customers arriving
- Number of defective products
- Number of emails received
- Number of goals scored

A discrete probability distribution tells us:

> **What values can occur, and how likely is each value?**

The main discrete probability distributions we will study are:

1. **Bernoulli Distribution**
2. **Binomial Distribution**
3. **Poisson Distribution**


### 2. Big Picture



Different real-world situations require different probability distributions.

####  Bernoulli Distribution

One trial with only two possible outcomes:

**Success / Failure**

Example:
- Customer clicks / does not click
- Product is defective / not defective
- Loan approved / rejected

####  Binomial Distribution

Multiple independent Bernoulli trials.

Example:

> Out of 100 customers, how many will click an advertisement?

####  Poisson Distribution

Counts how many times an event occurs within a fixed interval.

Example:

> How many customers arrive at a store in one hour?

####  Remember

**Bernoulli → One trial**

**Binomial → Fixed number of trials**

**Poisson → Number of events in a fixed interval**


### 3. Bernoulli Distribution


#### Concept


The **Bernoulli Distribution** is the simplest discrete probability distribution.

It models a **single trial** that has only **two possible outcomes**.

We usually represent them as:

- `1` → Success
- `0` → Failure

Examples:

- Coin → Heads / Tails
- Email → Opened / Not opened
- Product → Defective / Not defective
- Customer → Purchased / Did not purchase
- Exam → Pass / Fail

The important condition is:

> **There must be exactly one trial and two possible outcomes.**

#### Intuition


Imagine flipping a coin once.

There are only two possible outcomes:

**Heads → Success → 1**

**Tails → Failure → 0**

Suppose the probability of success is `p`.

Then the probability of failure is:

`1 - p`

So the distribution is:

| X | Probability |
|---|---:|
| 0 (Failure) | 1 - p |
| 1 (Success) | p |

For a fair coin:

p = 0.5

Therefore:

P(X = 1) = 0.5

P(X = 0) = 0.5

####  Formula


The probability mass function of a Bernoulli random variable is:

$$
P(X=x)=p^x(1-p)^{1-x}
$$

where:

- `X` can be `0` or `1`
- `p` = probability of success
- `1-p` = probability of failure

For `X = 1`:

$$
P(X=1)=p
$$

For `X = 0`:

$$
P(X=0)=1-p
$$

### 4. Python Implementation




```python
import numpy as np

p = 0.7

values = np.array([0, 1])
probabilities = np.array([1 - p, p])

print("Failure probability:", probabilities[0])
print("Success probability:", probabilities[1])
```

A Bernoulli distribution has only two possible values:

**0 → Failure**

**1 → Success**

The height of each bar represents its probability.

```python
import matplotlib.pyplot as plt

plt.bar(values, probabilities)

plt.xticks([0, 1], ["Failure (0)", "Success (1)"])
plt.ylabel("Probability")
plt.xlabel("Outcome")
plt.title("Bernoulli Distribution")

plt.show()
```

### 8. Bernoulli vs Binomial

The easiest way to distinguish them is by the **number of trials**.

#### Bernoulli

**One trial**

Example:

> Will this customer purchase?

Possible result:

**0 or 1**

#### Binomial

**Multiple fixed trials**

Example:

> Out of 100 customers, how many will purchase?

Possible results:

**0, 1, 2, ..., 100**

So:

**Bernoulli = one binary trial**

**Binomial = multiple independent Bernoulli trials**

#### Key Takeaways

- It has exactly **two possible outcomes**.
- Outcomes are usually represented as `0` and `1`.
- `1` usually represents success.
- `0` usually represents failure.
- `p` represents the probability of success.
- `1-p` represents the probability of failure.
- Bernoulli is the foundation of the **Binomial Distribution**.
