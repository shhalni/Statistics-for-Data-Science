## Sampling

### 1. Introduction

Sampling is the process of selecting a subset of observations from a larger population to study its characteristics.

A **population** is the entire group we want to study, while a **sample** is a smaller group selected from that population.

Sampling helps us understand a population without always analysing every individual.

Examples:

- Selecting students to study their academic performance
- Surveying customers about product satisfaction
- Testing a sample of manufactured products
- Studying a sample of patients to understand a disease
- Analysing a sample of transactions to detect fraud


### Why do we need Sampling

Sampling is useful when studying the entire population is difficult, expensive, or time-consuming.

#### 1. Saves Time and Cost

Collecting data from an entire population may require significant time and resources.

#### 2. Efficient Analysis

A smaller sample can make data exploration and analysis faster when working with very large datasets.

#### 3. Population Estimation

Sample statistics can be used to estimate population parameters.

#### 4. Statistical Inference

Sampling helps us understand uncertainty and forms the foundation of confidence intervals and hypothesis testing.

#### Example

Suppose a college has 10,000 students.

Instead of surveying every student, we select 500 students and calculate their average math score.

We can use this sample to estimate the average score of the larger population, provided the sampling method is appropriate.

**Important:** If the entire population is already available and can be analysed efficiently, sampling may not be necessary.


### Population and Sample


#### Population

A population is the entire group of individuals or observations we want to study.

#### Sample

A sample is a subset of individuals selected from the population.

#### Population Parameter

A population parameter is a numerical value describing the population.

Examples:
- Population mean (μ)
- Population variance (σ²)
- Population standard deviation (σ)

#### Sample Statistic

A sample statistic is a numerical value calculated from a sample.

Examples:
- Sample mean (x̄)
- Sample variance (s²)
- Sample standard deviation (s)

#### Example

Suppose a dataset contains the math scores of 1,000 students.

- Population: All 1,000 students in the dataset
- Sample: 100 selected students
- Population mean: Mean math score of all 1,000 students
- Sample mean: Mean math score of the 100 selected students

The sample mean may differ from the population mean because the sample contains only a subset of observations.

### Population and Sample - Formula

#### Population Mean

The population mean is calculated using all observations in the population.

$$
\mu = \frac{\sum_{i=1}^{N} x_i}{N}
$$

#### Sample Mean

The sample mean is calculated using observations selected from the population.

$$
\bar{x} = \frac{\sum_{i=1}^{n} x_i}{n}
$$

Where:

- μ = Population mean
- x̄ = Sample mean
- N = Population size
- n = Sample size
- xᵢ = Individual observation

The sample mean is commonly used to estimate the population mean

### Census vs Sampling

#### 1. Census

A census collects data from **every member of the population**.

Example: Collecting math scores from all 1,000 students in a college.

#### 2. Sampling

Sampling collects data from **only a subset of the population** to understand the larger population.

Example: Selecting 100 students from the college and analysing their math scores.

#### Key Differences

| Census | Sampling |
|---|---|
| Includes the entire population | Includes only a subset |
| Usually requires more time and resources | Usually requires less time and resources |
| No sampling error | Sampling error may occur |
| Useful when complete information is needed | Useful when studying everyone is impractical |

**Important:** A census can still contain non-sampling errors, such as incorrect measurements or missing responses. Sampling is often preferred when the population is large and a representative sample can provide reliable insights.

```python
# Census: use all available students
census = df

# Sampling: randomly select 100 students
sample = df.sample(n=100, random_state=15)

print("Total students:", len(census))
print("Sample size:", len(sample))

print("\nCensus mean math score:", census["math score"].mean())
print("Sample mean math score:", sample["math score"].mean())
```

The census uses every available student, while sampling uses only 100 students.

The sample mean may differ from the census mean because the sample contains only part of the population. This difference illustrates sampling variability.

### Probability vs. Non-Probability Sampling

Sampling methods are divided into two major categories.

#### 1. Probability Sampling

Every population member has a **known, non-zero probability of selection**.

Examples:
- Simple Random Sampling
- Systematic Sampling
- Stratified Sampling
- Cluster Sampling
- Multistage Sampling

**Intuition:** Selection follows a random process, which supports statistical inference about the population when the sampling design is appropriate.

#### 2. Non-Probability Sampling

Selection does not use a random mechanism that gives every population member a known selection probability.

Examples:
- Convenience Sampling
- Purposive Sampling
- Quota Sampling
- Snowball Sampling

**Intuition:** Participants are selected based on accessibility, specific requirements, group quotas, or referrals.

#### Key Difference

Probability sampling is generally better suited to estimating population characteristics and quantifying sampling uncertainty. Non-probability sampling can be useful when random selection is impractical or when studying a specific group, but it can introduce selection bias.

**Remember:** Random selection does not automatically guarantee a perfectly representative sample, and non-probability sampling is not automatically useless.

```python
# Probability sampling: simple random sample
random_sample = df.sample(n=100, random_state=15)

# Non-probability sampling: select the first 100 rows
convenience_sample = df.head(100)

print("Random sample mean:",
      random_sample["math score"].mean())

print("Convenience sample mean:",
      convenience_sample["math score"].mean())
```

- The random sample selects students through a random process.
- The convenience sample selects the first 100 rows because they are readily available.

Compare their means to see how different selection methods can produce different results. One comparison alone does not prove that either sample is biased; the selection process and population structure also matter.

### Simple Random Sampling

Simple random sampling is a probability sampling method in which every individual has an equal chance of being selected.

#### Example

Suppose a dataset contains 1,000 students.

We randomly select 100 students to study their math scores.

Each student has an equal chance of being selected.

#### When to Use It

- When we want a general sample of the population.
- When no specific subgroup needs guaranteed representation.
- When a suitable list of population members is available.

#### Advantages

- Easy to understand and implement.
- Reduces selection bias when properly conducted.

#### Limitation

A small random sample may not represent every subgroup adequately.

```python
# Select 100 students randomly

sample_random = df.sample(
    n=100,
    random_state=15
)

print("Sample Shape:", sample_random.shape)

print("\nFirst Five Rows:")
print(sample_random.head())

print("\nSample Mean:")
print(sample_random["math score"].mean())
```

#### Interpretation

The histogram compares the math score distributions of the full dataset and the selected sample.

If the distributions look similar, the sample may capture some important characteristics of the full dataset.

However, visual similarity alone does not prove that a sample is representative.

The sample mean and other statistics can differ due to sampling variability.

### Systematic Sampling

Systematic sampling selects individuals at regular intervals from an ordered population after choosing a starting point.

The sampling interval is calculated as:

$$
k = \frac{N}{n}
$$

Where:

- N = Population size
- n = Desired sample size
- k = Sampling interval

#### Example

Suppose there are 1,000 students and we want a sample of 100 students.

$$
k = \frac{1000}{100} = 10
$$

We select every 10th student after choosing a starting position.

#### When to Use It

- When observations are arranged in a list.
- When regular-interval selection is convenient.
- When the ordering does not introduce a repeating pattern that could bias selection.

#### Limitation

If the data contains a repeating pattern, systematic sampling may produce a biased sample.

```python
import numpy as np

N = len(df)
n = 100

# Calculate the sampling interval

k = N / n

# Choose a random starting position

rng = np.random.default_rng(42)
start = rng.uniform(0, k)

# Select n observations at regular intervals

indices = np.floor(
    start + np.arange(n) * k
).astype(int)

sample_systematic = df.iloc[indices]

print("Sample Shape:", sample_systematic.shape)
print("Sample Mean:", sample_systematic["math score"].mean())
```

#### Interpretation

Systematic sampling selects observations at regular intervals after choosing a starting point.

The sample mean can differ from the full dataset mean because the selected observations form only a subset.

Before using systematic sampling, we should check whether the order of the dataset contains patterns that could affect the results.

### Stratified Sampling

Stratified sampling divides a population into meaningful groups called **strata** and randomly selects observations from each group.

#### Example

Suppose the Student Performance dataset contains students from different gender categories.

We divide the students into gender groups and select a sample from each group.

This helps ensure that every gender category is represented.

#### When to Use It

- When important subgroups need to be represented.
- When groups differ in characteristics relevant to the study.
- When we want to estimate overall or subgroup statistics.

#### Advantages

- Ensures representation from each defined stratum.
- Can improve estimation precision when strata are meaningfully different.

#### Limitation

We need a suitable variable for creating the strata.

Stratified sampling does not automatically remove every source of bias.

```python
# Select 20% of students from each gender category

sample_stratified = (
    df.groupby("gender", group_keys=False)
      .sample(frac=0.20, random_state=42)
)

print("Sample Shape:", sample_stratified.shape)

print("\nGender Counts in Full Dataset:")
print(df["gender"].value_counts())

print("\nGender Counts in Stratified Sample:")
print(sample_stratified["gender"].value_counts())

print("\nStratified Sample Mean:")
print(sample_stratified["math score"].mean())
```

#### Interpretation

The code selects approximately 20% of the observations from each gender category.

This preserves representation from each defined group.

The total sample size may differ from a fixed target because the sample size is calculated separately for each group.

If the objective is to preserve each group's population proportion, proportional stratified sampling is useful.

If the objective is to compare groups, we may deliberately select equal or different sample sizes from each group.

```python
population_proportions = (
    df["gender"].value_counts(normalize=True)
)

sample_proportions = (
    sample_stratified["gender"].value_counts(normalize=True)
)

comparison = pd.DataFrame({
    "Full Dataset Proportion": population_proportions,
    "Sample Proportion": sample_proportions
})

print(comparison)
```

### Cluster sampling

Cluster sampling divides a population into natural groups called clusters.

Researchers randomly select some clusters and study the individuals within the selected clusters.

#### Example

A university has 50 classrooms.

Instead of randomly selecting individual students from every classroom, a researcher randomly selects five classrooms and studies the students in those classrooms.

Here:

- Population: All students in the university
- Clusters: Classrooms
- Selected clusters: Five classrooms
- Sample: Students belonging to the selected classrooms

#### When to Use It

- When individuals naturally belong to groups.
- When collecting data from selected groups is easier or cheaper.
- When studying geographically distributed populations.

#### Limitation

Selected clusters may differ from the overall population.

Cluster sampling requires genuine cluster identifiers, such as classroom, school, village, or branch.

**Dataset note:** If your Student Performance dataset does not contain a suitable cluster identifier, understand this method conceptually rather than inventing clusters.

### Multistage sampling

Multistage sampling selects a sample through **two or more stages**, usually by sampling groups first and then selecting individuals within those groups.

#### Example

Suppose a researcher wants to study student performance across a state.

1. Stage 1: Randomly select districts.
2. Stage 2: Randomly select colleges within the selected districts.
3. Stage 3: Randomly select students within those colleges.

#### Intuition

Instead of randomly selecting students from the entire state, the researcher narrows the selection through several stages.

#### When Is It Useful?

- Large populations spread across different locations
- National or state-level surveys
- Research involving schools, colleges, hospitals, or villages

**Difference from cluster sampling:** Cluster sampling often selects clusters and studies all members of the selected clusters. Multistage sampling can select clusters and then sample members within them.

```python
# Demonstration only: create artificial classroom groups.
# Replace these with real district, college, and student IDs
# when working with an actual multistage sampling dataset.

data = df.copy()

data["classroom_id"] = (
    np.arange(len(data)) // 30
)

# Stage 1: randomly select classrooms
selected_classrooms = data["classroom_id"].drop_duplicates().sample(
    n=min(3, data["classroom_id"].nunique()),
    random_state=42
)

stage_1 = data[
    data["classroom_id"].isin(selected_classrooms)
]

# Stage 2: randomly select students within the selected classrooms
multistage_sample = (
    stage_1.groupby("classroom_id", group_keys=False)
    .sample(frac=0.5, random_state=42)
)

print("Original dataset size:", len(data))
print("Selected classrooms:", len(selected_classrooms))
print("Final multistage sample size:", len(multistage_sample))
```

The classroom IDs above are artificially created for demonstration. Therefore, this code illustrates the mechanics of multistage sampling, not a genuine classroom-based sampling design. Real research should use actual group identifiers and a clearly defined selection process.

### Purposive Sampling

Purposive sampling is a non-probability sampling method in which the researcher deliberately selects participants who meet a particular criterion or possess relevant characteristics.

#### Example

A researcher studying high academic achievement selects students who scored at least 90 in mathematics.

#### Intuition

The researcher is not trying to select a random cross-section of all students. Instead, they specifically want to study a particular group.

#### When Is It Useful?

- Expert interviews
- Case studies
- Research involving specialised knowledge
- Studies focused on a specific group or characteristic

**Limitation:** Findings from a purposive sample may not represent the entire population.

```python
# Select students who meet a specific research criterion
purposive_sample = df[
    df["math score"] >= 90
]

print("Original dataset size:", len(df))
print("Purposive sample size:", len(purposive_sample))

purposive_sample[
    ["math score", "reading score", "writing score"]
].head()
```

This sample deliberately focuses on students with math scores of 90 or above. It is suitable for studying high-performing students, but it should not be used to estimate the average performance of all students without additional justification.

### Convenience Sampling

Convenience sampling is a non-probability sampling method in which individuals are selected because they are easy to access.

#### Example

A researcher surveys only students sitting nearby instead of selecting students randomly from the entire college.

#### Advantages

- Quick and easy to conduct.
- Requires fewer resources.
- Useful for preliminary exploration.

#### Limitations

- May introduce selection bias.
- May not represent the entire population.
- Results may not generalize well to the wider population.

Convenience sampling should be used cautiously when making conclusions about a larger population.

```python
# Illustrative convenience sample:
# Select the first 100 rows.

sample_convenience = df.head(100)

print("Sample Shape:", sample_convenience.shape)
print("Sample Mean:", sample_convenience["math score"].mean())
```

#### Important Note

Selecting the first 100 rows is only an illustration of convenience-like selection.

It is not necessarily a true convenience sample, and it may not represent how real-world convenience sampling is performed.

If the dataset is ordered by score, gender, or another characteristic, selecting the first 100 rows may produce a misleading estimate.

The order and selection process must be considered before interpreting the results.

### Quota Sampling

Quota sampling is a non-probability sampling method in which the researcher sets a target number of participants for particular groups and continues selecting participants until each quota is filled.

#### Example

A researcher wants a sample of 100 students:
- 50 male students
- 50 female students

The researcher fills both quotas but does not randomly select students within each group.

#### Intuition

The sample must contain a specified number of people from each group, but the selection within those groups is not necessarily random.

#### Quota Sampling vs. Stratified Sampling

- **Quota sampling:** Participants within each group are usually selected non-randomly.
- **Stratified sampling:** Participants are randomly selected from each stratum.

Quota sampling can ensure that selected groups are represented in the sample, but it does not automatically make the sample representative of the population.

```python
# Example: target 50 students from each gender.
# Taking the first available rows within each group is
# a non-random selection method for this demonstration.

gender_counts = df["gender"].value_counts()
print("Available students by gender:")
print(gender_counts)

target_per_gender = 50
quota_samples = []

for gender in ["male", "female"]:
    group = df[df["gender"] == gender]

    selected = group.head(
        min(target_per_gender, len(group))
    )

    quota_samples.append(selected)

quota_sample = pd.concat(quota_samples)

print("\nQuota sample counts:")
print(quota_sample["gender"].value_counts())

print("\nTotal quota sample size:", len(quota_sample))
```

The code attempts to select 50 students from each gender group. If a group contains fewer than 50 students, the code selects all available students from that group.

Because the first available rows are selected, this is a simple demonstration of non-random quota filling. A real study should define its quotas according to the research objective.

### Snowball Sampling

Snowball sampling is a non-probability sampling method in which existing participants refer or recruit other eligible participants.

#### Example

A researcher studying a rare professional community interviews one member. That participant introduces the researcher to other members, who may then refer additional participants.

#### Intuition

The sample grows through participant referrals, much like a snowball growing as it rolls.

#### When Is It Useful?

- Hard-to-reach populations
- Rare or specialised communities
- Studies where participants are difficult to identify through ordinary sampling methods

#### Limitations

- Participants may refer people from similar social networks.
- Some groups may be overrepresented while others are missed.
- Selection probabilities are generally unknown, limiting population-wide generalisation.

```python
# A small, illustrative referral network.
# Each person refers other people who may join the study.

referrals = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["E", "F"],
    "D": ["G"],
    "E": [],
    "F": ["H"],
    "G": [],
    "H": []
}

# Start with one participant
sampled_people = {"A"}
frontier = ["A"]

# Limit the number of referral rounds
for _ in range(3):
    next_frontier = []

    for person in frontier:
        for referral in referrals.get(person, []):
            if referral not in sampled_people:
                sampled_people.add(referral)
                next_frontier.append(referral)

    frontier = next_frontier

print("Participants recruited:")
print(sorted(sampled_people))

print("Total participants:", len(sampled_people))
```

The simulation starts with participant A and follows referrals for three rounds. Each newly recruited participant can introduce additional participants.

This demonstrates how a sample can grow through referrals. It is a simulated network, not actual data collected from people.

### Comparison of Sampling Methods


| Method | Category | Selection approach |
|---|---|---|
| Simple Random | Probability | Randomly select individuals |
| Systematic | Probability | Select individuals at a fixed interval after a random start |
| Stratified | Probability | Randomly select from each stratum |
| Cluster | Probability | Randomly select groups or clusters |
| Multistage | Probability | Sample through multiple stages |
| Convenience | Non-probability | Select readily available participants |
| Purposive | Non-probability | Select participants meeting specific criteria |
| Quota | Non-probability | Fill predefined group quotas non-randomly |
| Snowball | Non-probability | Recruit through participant referrals |

**Key takeaway:** Choose a sampling method based on the research question, population, available resources, and the degree to which you need to generalise findings to the population.

```python
N = len(df)
n = 100
k = N / n

rng = np.random.default_rng(42)
start = rng.uniform(0, k)

indices = np.floor(
    start + np.arange(n) * k
).astype(int)

sample_systematic = df.iloc[indices]

print("Sample shape:", sample_systematic.shape)
print("Sample mean:", sample_systematic["math score"].mean())

# Full dataset mean

full_mean = df["math score"].mean()

# Calculate means for the different samples

sample_means = {
    "Full Dataset": full_mean,
    "Simple Random": sample_random["math score"].mean(),
    "Systematic": sample_systematic["math score"].mean(),
    "Stratified": sample_stratified["math score"].mean(),
    "Convenience-like": sample_convenience["math score"].mean()
}

comparison_df = pd.DataFrame(
    list(sample_means.items()),
    columns=["Sampling Method", "Mean Math Score"]
)

comparison_df["Absolute Difference"] = (
    comparison_df["Mean Math Score"] - full_mean
).abs()

print(comparison_df)
```

#### Interpretation

The comparison table shows the mean math score obtained from the full dataset and each sample.

The absolute difference measures how far each sample mean is from the full dataset mean.

A smaller difference means the sample mean is closer to the full dataset mean for this particular dataset and sample.

However, a single comparison does not prove that one sampling method is always better.

Different random samples can produce different estimates. Sampling methods should be chosen based on the research objective, population structure, and potential sources of bias.

### ML Connection

Sampling is important in machine learning because models are trained and evaluated using selected observations.

#### 1. Train-Test Split

A dataset is divided into training and testing subsets to evaluate how well a model performs on unseen data.

#### 2. Stratified Train-Test Split

For classification problems, stratification can help preserve class proportions in training and testing sets.

#### 3. Random Sampling

Randomly selecting observations can be useful when creating subsets for exploration or experimentation.

#### 4. Sampling Bias

If the training data does not represent the population where the model will be used, the model may perform poorly on that population.

#### Important Distinction

Sampling methods are not the same as train-test splitting.

Sampling selects observations for a particular purpose, while train-test splitting separates data for model training and evaluation.

For time-series data, random splitting may be inappropriate because the chronological order of observations matters.
