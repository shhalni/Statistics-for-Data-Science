## Uniform distribution

### 1. Introduction

A **Uniform Distribution** is a probability distribution where all values within a specified range have the **same probability density**.

In simple words:

> **Every part of the range is equally likely.**

For example, suppose we randomly select a number between 0 and 10.

If the selection is truly uniform, values around:

- 1
- 5
- 7
- 9

are equally likely.

There is no preferred region.

A Uniform Distribution can be:

- **Discrete** — individual values have equal probability
- **Continuous** — every region of equal width has equal probability density

In this section, we focus mainly on the **continuous Uniform Distribution**.


### 2. Intuition


Imagine a ruler from 0 to 10.

```text
0 ───────────────────────────── 10

Suppose we randomly choose a point on this ruler.

There is no reason for the point to prefer one part of the ruler over another.

So:

0–2 is just as likely as 4–6
6–8 is just as likely as 8–10

as long as the intervals have the same width.

This is why it is called Uniform — probability is distributed evenly across the range.

Mental Model

Uniform = Equal likelihood across the range

#### Shape of a Continuous Uniform Distribution

The PDF of a continuous Uniform Distribution is a **flat horizontal line**.

For example, between `a` and `b`:

```text
Density
   ↑
   │       ┌─────────────────┐
   │       │                 │
   │       │                 │
   │───────┘                 └────────→ X
           a                 b

The density is constant between a and b.

Outside this range, the density is 0.

The area under the curve represents probability.

Because the height is constant, equal-width intervals have equal probabilities.

### 3. Formula

For a continuous Uniform Distribution between `a` and `b`:

$$
f(x)=\frac{1}{b-a}
$$

for:

$$
a \leq x \leq b
$$

where:

- `a` = lower limit
- `b` = upper limit

The probability of a range can be found using:

$$
P(c \leq X \leq d)=\frac{d-c}{b-a}
$$

This makes sense because probability depends on the **length of the interval**.

### Example

If:

`X ~ Uniform(0, 10)`

Then:

$$
P(2 \leq X \leq 5)
=
\frac{5-2}{10-0}
=
\frac{3}{10}
=
0.3
$$

So the probability is **30%**.

````python
from scipy.stats import uniform

a = 0
b = 10

probability = uniform.cdf(5, loc=a, scale=b-a) - uniform.cdf(2, loc=a, scale=b-a)

print("Probability:", probability)
print("Percentage:", probability * 100)
````

#### Generating Uniform Random Values

We can use NumPy to generate random values uniformly between two limits.

For example, we can generate random numbers between 0 and 10.

````python
import numpy as np

np.random.seed(42)

values = np.random.uniform(0, 10, 1000)

print(values[:10])
````

#### Visualization

If we generate many random values from a Uniform Distribution, the histogram should look approximately **flat**.

This happens because values are spread relatively evenly across the range.

````python
import matplotlib.pyplot as plt

plt.hist(values, bins=20, density=True)

plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Continuous Uniform Distribution")

plt.show()
````

### 4. Real-World Example

Suppose a system randomly generates a number between 0 and 1, and every value in that range is equally likely.

This follows a Continuous Uniform Distribution:

$$
X \sim Uniform(0,1)
$$

For example:

$$
P(0.2 \leq X \leq 0.5)
=
\frac{0.5-0.2}{1-0}
=
0.3
$$

Therefore, there is a **30% probability** that the generated value falls between 0.2 and 0.5.

### 5. Business Applications

Uniform Distribution can be used when outcomes are assumed to be equally likely across a defined range.

Examples:

- Random sampling
- Simulation
- Random number generation
- Modeling equally likely arrival times within a short interval
- Monte Carlo simulations
- Testing systems with randomly generated inputs

In real business data, a perfectly uniform distribution is relatively uncommon, but the distribution is very useful for **simulation and modeling**.

### 6. ML Connection

Uniform Distribution is commonly used in machine learning for:

- Random initialization
- Simulation
- Generating synthetic data
- Random sampling
- Hyperparameter initialization

For example, a model may initialize parameters by randomly selecting values from a specified range.

The important idea is:

> **Uniform Distribution provides a simple way to generate values without favoring one part of the chosen range.**

#### Key Takeaways

- Uniform Distribution gives equal probability density across a specified range.
- It has a flat PDF.
- It can be discrete or continuous.
- In a continuous Uniform Distribution, equal-width intervals have equal probabilities.
- The probability of a range depends on the length of that range.

For `X ~ Uniform(a,b)`:
