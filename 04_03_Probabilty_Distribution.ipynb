## Probability Distribution

### 1. Introduction

A **probability distribution** describes all the possible values that a random variable can take and the probability associated with each value.

In simple terms, it answers two questions:

1. **What values can the random variable take?**
2. **How likely is each value?**

For example, when rolling a fair six-sided die:

$$
X \in \{1,2,3,4,5,6\}
$$

Each value has a probability of:

$$
P(X=x)=\frac{1}{6}
$$

Therefore, the probability distribution tells us how the probability is distributed across the possible values of \(X\).


### 2. Intuition


Think of a probability distribution as a **map of uncertainty**.

A random variable tells us **what can happen**.

A probability distribution tells us **how likely each possibility is**.

For example:

$$
X=\text{number obtained when rolling a die}
$$

The random variable tells us:

$$
X \in \{1,2,3,4,5,6\}
$$

The probability distribution tells us:

$$
P(X=1)=\frac16
$$

$$
P(X=2)=\frac16
$$

$$
\vdots
$$

$$
P(X=6)=\frac16
$$

Therefore:

$$
\text{Random Variable} \rightarrow \text{Possible Values}
$$

$$
\text{Probability Distribution} \rightarrow \text{Likelihood of Those Values}
$$

### 3. Example - Fair Die

Consider a fair six-sided die.

Let:

$$
X=\text{number appearing on the die}
$$

The possible values are:

$$
X \in \{1,2,3,4,5,6\}
$$

Since the die is fair, every outcome has an equal probability.

| \(X\) | \(P(X=x)\) |
|---:|---:|
| 1 | \(1/6\) |
| 2 | \(1/6\) |
| 3 | \(1/6\) |
| 4 | \(1/6\) |
| 5 | \(1/6\) |
| 6 | \(1/6\) |

The probabilities must add up to 1:

$$
\frac16+\frac16+\frac16+\frac16+\frac16+\frac16=1
$$

Therefore, this table represents a valid probability distribution.

### 4. Properties

For a valid discrete probability distribution, two important conditions must be satisfied.

#### 1. Probability must be between 0 and 1

For every possible value:

$$
0 \leq P(X=x) \leq 1
$$

A probability cannot be negative or greater than 1.

#### 2. Total probability must equal 1

The probabilities of all possible outcomes must add up to 1:

$$
\sum_x P(X=x)=1
$$

This is because all possible outcomes together represent the complete set of possibilities.

### 5. Python Implementation




We can represent a probability distribution using Python lists or NumPy arrays.

For a fair die, each outcome has probability \(1/6\).

#### dice

```python
import numpy as np

outcomes = np.array([1, 2, 3, 4, 5, 6])
probabilities = np.array([1/6] * 6)

print("Outcomes:", outcomes)
print("Probabilities:", probabilities)
print("Total Probability:", probabilities.sum())
```

### 6. Verifying a Probability Distribution


We can check whether a probability distribution is valid by verifying:

1. Every probability is between 0 and 1.
2. The total probability equals 1.

```python
valid_probabilities = np.all(
    (probabilities >= 0) & (probabilities <= 1)
)

total_probability = probabilities.sum()

print("All probabilities valid:", valid_probabilities)
print("Total probability:", total_probability)
print("Is valid distribution:", valid_probabilities and np.isclose(total_probability, 1))
```

### 7. Visualization

For a discrete random variable, a **bar chart** is appropriate because the possible values are separate and countable.

For a fair die, all bars should have the same height because every outcome has the same probability.

```python
import matplotlib.pyplot as plt

plt.bar(outcomes, probabilities)

plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Probability Distribution of a Fair Die")

plt.xticks(outcomes)

plt.show()
```

### 8. Unequal probabilities

Not every probability distribution gives equal probability to every outcome.

For example, suppose a random variable \(X\) has the following distribution:

| \(X\) | \(P(X=x)\) |
|---:|---:|
| 0 | 0.1 |
| 1 | 0.2 |
| 2 | 0.4 |
| 3 | 0.3 |

The probabilities add up to:

$$
0.1+0.2+0.4+0.3=1
$$

This is a valid probability distribution.

Here, \(X=2\) is the most likely outcome because:

$$
P(X=2)=0.4
$$

A probability distribution therefore does not necessarily mean that all outcomes are equally likely.

```python
outcomes = np.array([0, 1, 2, 3])
probabilities = np.array([0.1, 0.2, 0.4, 0.3])

plt.bar(outcomes, probabilities)

plt.xlabel("Value of X")
plt.ylabel("Probability")
plt.title("Discrete Probability Distribution")

plt.show()
```

### 9. Connection to Expected Value


A probability distribution tells us the possible values of a random variable and their probabilities.

Once we have this distribution, we can calculate the **Expected Value**.

For a discrete random variable:

$$
E[X]=\sum_x xP(X=x)
$$

Using the distribution:

| \(X\) | \(P(X=x)\) |
|---:|---:|
| 0 | 0.1 |
| 1 | 0.2 |
| 2 | 0.4 |
| 3 | 0.3 |

We calculate:

$$
E[X]
=
0(0.1)+1(0.2)+2(0.4)+3(0.3)
$$

$$
=0+0.2+0.8+0.9
$$

$$
=1.9
$$

Therefore:

$$
E[X]=1.9
$$

This shows how probability distributions connect directly to Expected Value.

```python
expected_value = np.sum(outcomes * probabilities)

print("Expected Value:", expected_value)
```

### 10. Probability vs Probability Distribution


These two concepts are related but not exactly the same.

#### Probability

Probability usually describes the likelihood of a particular event.

For example:

$$
P(X=4)=\frac16
$$

This tells us the probability of getting 4 when rolling a fair die.

#### Probability Distribution

A probability distribution describes the probabilities of **all possible values** of the random variable.

For the die:

$$
P(X=1),P(X=2),...,P(X=6)
$$

Therefore:

> **Probability → likelihood of an event or outcome**

> **Probability Distribution → complete picture of probabilities across possible values**

### 🧠 Theoretical vs Empirical Distribution

#### Theoretical Probability Distribution

A theoretical distribution is defined mathematically.

Examples:

- Bernoulli
- Binomial
- Poisson
- Normal
- Uniform

#### Empirical Distribution

An empirical distribution is constructed from observed real-world data.

For example, the Titanic age distribution is based on the actual ages recorded in the dataset.

Therefore:

$$
\text{Theoretical Distribution}
\rightarrow
\text{Mathematical Model}
$$

$$
\text{Empirical Distribution}
\rightarrow
\text{Observed Data}
$$
