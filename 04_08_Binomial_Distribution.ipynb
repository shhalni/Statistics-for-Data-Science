## Binomial distribution

### 1. Introduction

The **Binomial Distribution** is used to find the probability of getting a certain number of **successes** in a fixed number of independent trials.

It is based on repeated **Bernoulli trials**.

For a Binomial Distribution:

- There is a fixed number of trials → `n`
- Each trial has only two outcomes → Success / Failure
- Probability of success remains the same → `p`
- Trials are independent
- We count the total number of successes → `X`

#### Simple Example

Suppose an advertisement has a 20% chance of being clicked.

If it is shown to **10 customers**, we can ask:

> What is the probability that exactly 3 customers click the advertisement?

This is a **Binomial Distribution** problem.


### 2. Intuition


Remember Bernoulli:

**One customer → Click / No Click**

Now repeat the same experiment:

**10 customers → Click / No Click for each customer**

We are not interested in the exact sequence of clicks.

We only want to know:

> **How many customers clicked?**

The answer could be:

`0, 1, 2, 3, ..., 10`

That number of successes is the **Binomial random variable**.

### Mental Model

**Bernoulli → One binary trial**

**Binomial → Many binary trials + count the successes**mber of events in a fixed interval**


### 3. Conditions


A situation follows a Binomial Distribution when these conditions are satisfied:

1. **Fixed number of trials**
   - We decide the number of trials beforehand.

2. **Two possible outcomes**
   - Each trial has success or failure.

3. **Independent trials**
   - The result of one trial does not affect another.

4. **Constant probability of success**
   - The probability `p` remains the same for every trial.

5. **Count the successes**
   - We are interested in the total number of successes.

### 4. Formula


The probability of getting exactly `k` successes in `n` trials is:

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
$$

Where:

- `n` = total number of trials
- `k` = number of successes
- `p` = probability of success
- `1-p` = probability of failure
- `n-k` = number of failures
- `nCk` = number of ways to arrange the successes and failures

The formula essentially means:

**Ways to arrange the successes × Probability of those successes and failures**

### 5. Simple Example


Suppose:

- Number of customers = `10`
- Probability of purchase = `0.2`
- We want exactly `3` purchases

Therefore:

`n = 10`

`k = 3`

`p = 0.2`

`1-p = 0.8`

We want:

$$
P(X=3)
$$

Using the Binomial formula:

$$
P(X=3)=\binom{10}{3}(0.2)^3(0.8)^7
$$

The result is approximately:

**0.2013 = 20.13%**

So there is approximately a **20.13% probability that exactly 3 out of 10 customers purchase the product.**

### 6. Python Implementation




```python
from scipy.stats import binom

n = 10
p = 0.2
k = 3

probability = binom.pmf(k, n, p)

print("Probability of exactly 3 successes:", probability)
print("Percentage:", probability * 100)
```

#### Probability of Different Numbers of Successes


Instead of asking only:

> "What is the probability of exactly 3 purchases?"

we can calculate the probability of getting:

- 0 purchases
- 1 purchase
- 2 purchases
- ...
- 10 purchases

This gives us the **complete Binomial probability distribution** for the number of purchases.

#### Binomial Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.2

x = np.arange(0, n + 1)
probabilities = binom.pmf(x, n, p)

plt.bar(x, probabilities)

plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")

plt.show()
```

### 7. Real-World Example

Suppose an e-commerce company sends an offer to 100 customers.

Historical data shows that each customer has a 10% probability of purchasing.

We can model the number of purchases using a Binomial Distribution if:

- There are exactly 100 customers
- Each customer either purchases or does not purchase
- Customers' decisions are treated as independent
- The purchase probability remains approximately 10%

We can then ask:

> What is the probability of getting exactly 8 purchases?

Or:

> What is the probability of getting at least 10 purchases?

This is where Binomial Distribution becomes useful in business analysis.

### 8. Business Applications

Binomial Distribution can be used when we count successes across a fixed number of trials.

Examples:

- Number of customers who purchase
- Number of users who click an advertisement
- Number of emails that are opened
- Number of defective products in a sample
- Number of successful loan applications
- Number of customers who churn
- Number of patients who respond to a treatment

### Key Question

> **"Out of a fixed number of trials, how many successes will occur?"**

### Key Takeaways
-   Binomial Distribution models the **number of successes in a fixed number of trials**.
- Each trial must have two possible outcomes.
- Trials should be independent.
- Probability of success remains constant.
- `n` = number of trials.
- `k` = number of successes.
- `p` = probability of success.

#### Mental Model

**Bernoulli → One trial**

**Binomial → Many Bernoulli trials + Count successes**
