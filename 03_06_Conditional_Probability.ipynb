## Conditional Probability

### 1. Introduction

**Conditional Probability** is the probability of an event occurring **given that another event has already occurred**.

The word **"given"** is the key indicator of conditional probability.

Unlike ordinary probability, conditional probability updates our calculation because we already know some additional information.

For example:

- Probability that a passenger survived **given** they were female.
- Probability that a patient has a disease **given** the test result is positive.
- Probability that a customer buys a product **given** they clicked an advertisement.

Conditional probability is widely used in statistics, machine learning, medical diagnosis, finance, and recommendation systems.ne learning.


### 2. Intuition


Imagine there are **100 students** in a class.

- 60 are boys.
- 40 are girls.

Suppose someone asks,

> "What is the probability of selecting a girl?"

You consider **all 100 students**.

Now suppose the question changes to

> "What is the probability of selecting a girl **given that the selected student is a CS student**?"

Now you **ignore all non-CS students** and calculate the probability only among CS students.

This is the main idea of conditional probability.

The condition changes the sample space.

Instead of looking at the entire population, we only look at the outcomes that satisfy the given condition.ake absolute predictions.


### 3. Mathematical Perspective


The conditional probability of event **A** given event **B** is

$$
P(A|B)=\frac{P(A\cap B)}{P(B)}
$$

where

- \(P(A|B)\) = Probability of A given B
- \(P(A \cap B)\) = Probability that both A and B occur
- \(P(B)\) = Probability that event B occurs

The denominator changes because we now consider only the outcomes where **B has already occurred**.efore,

$$
P(E)=\frac36=\frac12
$$

#### 🧩 Understanding the Formula

Suppose

- A = Passenger Survived
- B = Passenger is Female

Then

$$
P(\text{Survived}|\text{Female})
$$

means

> "Among female passengers, what is the probability that the passenger survived?"

Notice that we are **not dividing by the total number of passengers**.

Instead,

we divide only by the number of female passengers because that is our new sample space.

#### Example 1

A card is drawn from a standard deck.

Find the probability that the card is an **Ace given that it is a Spade**.

Sample Space after condition:

Only Spades

Total Spades = 13

Favourable Outcomes = Ace of Spades = 1

Therefore,

$$
P(Ace|Spade)=\frac1{13}
$$

#### Example 2

A die is rolled.

Find the probability of getting an **even number given that the number is greater than 3**.

Sample Space becomes

$$
\{4,5,6\}
$$

Even Numbers

$$
\{4,6\}
$$

Therefore

$$
P(Even|Greater\ Than\ 3)=\frac23
$$

### 4. Python Implementation




#### dice

### 5. Real Dataset Example (Titanic)



```python
import pandas as pd
df = pd.read_csv(r"C:\Users\WIN 10\Documents\stats and probability\Titanic-Dataset.csv")
df.head()
```

```python
#Probability that a passenger survived given they were female

female = df[df["Sex"]=="female"]

prob = (female["Survived"]==1).mean()

print(prob)
```

```python
#Probability that a passenger survived given they travelled in First Class
first = df[df["Pclass"]==1]

prob = (first["Survived"]==1).mean()

print(prob)
```

```python
#Probability that a passenger is female given they survived
survived = df[df["Survived"]==1]

prob = (survived["Sex"]=="female").mean()

print(prob)
```

### 7. ML Connection

Conditional probability is one of the most important concepts in machine learning.

Applications include:

- Naive Bayes Classifier
- Bayesian Networks
- Spam Detection
- Medical Diagnosis
- Fraud Detection
- Recommendation Systems

For example,

Naive Bayes calculates

> Probability that an email is spam **given** the words contained in the email.

Similarly,

Medical diagnosis predicts

> Probability that a patient has a disease **given** the observed symptoms.

Conditional probability forms the mathematical foundation of Bayes' Theorem, which is widely used in probabilistic machine learning models.
