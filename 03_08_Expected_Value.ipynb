## Expected Value

### 1. Introduction

**Expected Value (EV)** is the long-run average value we expect from a random experiment if the experiment is repeated many times.

It does not necessarily represent the value we will actually observe in a single trial.

Instead, it tells us the **average outcome we would expect over many repeated trials**.

Expected Value is widely used in probability, statistics, finance, insurance, business decision-making, and machine learning.

Expected Value is the **weighted average of all possible outcomes**, where each outcome is weighted by its probability of occurring.

For a discrete random variable \(X\), the expected value is:

$$
E[X] = \sum_{i=1}^{n} x_i P(X=x_i)
$$

where:

- \(E[X]\) = Expected value of \(X\)
- \(x_i\) = A possible outcome
- \(P(X=x_i)\) = Probability of that outcome

The probabilities of all possible outcomes must add up to 1:

$$
\sum_{i=1}^{n}P(X=x_i)=1
$$



### 2. Intuition


Imagine rolling a fair six-sided die.

The possible outcomes are:

$$
1,2,3,4,5,6
$$

Each outcome has the same probability:

$$
\frac{1}{6}
$$

The expected value is:

$$
E[X]
=
1\left(\frac16\right)
+
2\left(\frac16\right)
+
3\left(\frac16\right)
+
4\left(\frac16\right)
+
5\left(\frac16\right)
+
6\left(\frac16\right)
$$

Therefore:

$$
E[X]=3.5
$$

Notice that **3.5 is not actually a possible outcome of a single die roll**.

This is important.

Expected value does not mean:

> "What outcome will happen next?"

It means:

> "What average outcome would we expect over a large number of repeated trials?"

### 3. Python Implementation




#### dice

```python
outcomes = [1, 2, 3, 4, 5, 6]
probabilities = [1/6] * 6

expected_value = sum(
    x * p for x, p in zip(outcomes, probabilities)
)

print("Expected Value:", expected_value)
```

#### NumPy Implementation

```python
import numpy as np

outcomes = np.array([1, 2, 3, 4, 5, 6])
probabilities = np.array([1/6] * 6)

expected_value = np.sum(outcomes * probabilities)

print("Expected Value:", expected_value)
```

Expected value does not require all outcomes to have equal probabilities.

Suppose a game has the following possible outcomes:

| Outcome | Probability |
|---:|---:|
| ₹0 | 0.5 |
| ₹10 | 0.3 |
| ₹50 | 0.2 |

The expected value is:

$$
E[X]
=
0(0.5)
+
10(0.3)
+
50(0.2)
$$

$$
=0+3+10
$$

$$
=₹13
$$

Therefore, the expected value is **₹13**.

This means that if the game were repeated many times, the average payoff would approach ₹13 per game.

```python
outcomes = np.array([0, 10, 50])
probabilities = np.array([0.5, 0.3, 0.2])

expected_value = np.sum(outcomes * probabilities)

print("Expected Value:", expected_value)
```

```python
import matplotlib.pyplot as plt

outcomes = [0, 10, 50]
probabilities = [0.5, 0.3, 0.2]

plt.bar(outcomes, probabilities, width=5)

plt.xlabel("Outcome (₹)")
plt.ylabel("Probability")
plt.title("Possible Outcomes and Their Probabilities")

plt.show()
```

The expected value represents the **probability-weighted average** of the outcomes.

For example, an outcome with a high probability contributes more to the expected value than an outcome with a very low probability.

### Real Simulation

```python

import numpy as np

rolls = np.random.randint(1, 7, size=10000)

observed_average = rolls.mean()

print("Expected Value:", 3.5)
print("Observed Average:", observed_average)
```

### 4. Business Applications

Expected value is useful when businesses need to make decisions under uncertainty.

#### 🏦 Insurance

Insurance companies estimate expected claim costs.

For example:

- Probability of a claim
- Expected claim amount

These can be combined to estimate expected losses.

---

#### 💰 Investment

Investors can calculate expected returns by considering different possible outcomes and their probabilities.

For example:

- High return with low probability
- Moderate return with medium probability
- Loss with some probability

The expected return provides a probability-weighted average.

---

#### 🛒 E-commerce

A company can estimate the expected revenue from a customer by considering different possible purchasing behaviors.

---

#### 📢 Marketing

Expected value can help estimate the expected revenue from an advertising campaign based on possible customer responses.

---

#### 🎰 Risk Analysis

Businesses can compare different decisions by calculatin their expected financial outcomes.

#### #Key Takeaway

Expected value allows businesses to compare uncertain decisions using their probability-weighted average outcomes.g their probability-weighted average outcomes.

### 5. ML Connection


####  Loss Functions

Machine learning models often evaluate their performance using a loss function.

The model aims to minimize the expected loss over the data distribution.

Conceptually:

$$
\text{Expected Loss}
=
E[L(Y,\hat{Y})]
$$

where:

- \(Y\) = Actual value
- \(\hat{Y}\) = Predicted value
- \(L\) = Loss function

---

####  Decision Making

Expected values can also be used when choosing between actions under uncertainty.

For example, a model may estimate the probability of different outcomes and calculate the expected cost or benefit of each possible decision.

---

####  Reinforcement Learning

Expected rewards are fundamental to reinforcement learning.

An agent evaluates actions based on the expected future rewards they can produce.

####  Key Takeaway

Expected value provides a mathematical framework for making predictions and decisions when outcomes are uncertain.
