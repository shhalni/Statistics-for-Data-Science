# Probability

# 📖 Introduction

Probability is a branch of mathematics that measures the **likelihood or chance of an event occurring**. It provides a numerical value between **0 and 1**, where:

- **0** means the event is impossible.
- **1** means the event is certain.
- Values between **0 and 1** represent different levels of uncertainty.

In real-world scenarios, many events cannot be predicted with complete certainty. For example, we cannot say with confidence whether it will rain tomorrow or whether a customer will purchase a product. However, we can estimate the likelihood of these events using probability.

---

## Why is Probability Important?

Probability forms the foundation of **Statistics**, **Data Science**, and **Machine Learning**. It helps us:

- Analyze uncertainty in data.
- Make informed decisions using incomplete information.
- Build predictive models.
- Quantify risks and confidence in predictions.

Many machine learning algorithms use probability to estimate how likely an observation belongs to a particular class or outcome.

---

## 1. Random Experiment

### 1. Introduction

A **Random Experiment** is an experiment or process whose **exact outcome cannot be predicted with certainty**, even though all the possible outcomes are known beforehand.

When the experiment is repeated under identical conditions, the outcome may vary from one trial to another.

Unlike deterministic experiments, where the result is always the same, random experiments involve **uncertainty**.

Random experiments form the foundation of **Probability**, as probability measures the likelihood of different outcomes produced by these experiments.

---


### 2. Intuition


Imagine tossing a fair coin.

Before tossing, you know the possible outcomes are:

- Heads
- Tails

However, you cannot predict with certainty whether the next toss will result in Heads or Tails.

Similarly,

- Rolling a die
- Drawing a card from a deck
- Selecting a passenger from the Titanic dataset
- Predicting tomorrow's weather

are all examples of random experiments.

The uncertainty associated with these experiments is what probability helps us quantify.

---
#### 🌍 Examples

| Random Experiment | Possible Outcomes |
|-------------------|-------------------|
| Toss a Coin | Heads, Tails |
| Roll a Die | 1,2,3,4,5,6 |
| Draw a Card | Any one of the 52 cards |
| Weather Tomorrow | Rain, No Rain |
| Customer Purchase | Purchase, No Purchase |
| Student Exam Result | Pass, Fail |

---


### 3. Mathematical Perspective


A random experiment itself does not have a mathematical formula.

It defines the **sample space**, which is the set of all possible outcomes.

For a coin toss,

$$
S=\{H,T\}
$$

where

- H = Heads
- T = Tails

---

For rolling a die,

$$
S=\{1,2,3,4,5,6\}
$$

### 4. Python Implementation




#### Coin Toss Simulation

```python
import random

print("Coin Toss Simulation\n")

for i in range(10):
    outcome = random.choice(["Heads", "Tails"])
    print(f"Toss {i+1}: {outcome}")
```

#### Dice Roll Simulation

```python
import random

print("Dice Roll Simulation\n")

for i in range(10):
    outcome = random.randint(1,6)
    print(f"Roll {i+1}: {outcome}")
```


#### Card Selection

```python
import random

cards = ["Ace", "King", "Queen", "Jack"]

selected_card = random.choice(cards)

print("Selected Card:", selected_card)
```

### 5. Visualization



```python
import random
import matplotlib.pyplot as plt
from collections import Counter

# Simulate 1000 coin tosses
coin_tosses = [random.choice(["Heads", "Tails"]) for _ in range(1000)]

counts = Counter(coin_tosses)

plt.figure(figsize=(6,4))

plt.bar(counts.keys(), counts.values(), edgecolor="black")

plt.title("Coin Toss Simulation (1000 Tosses)")
plt.xlabel("Outcome")
plt.ylabel("Frequency")

plt.grid(axis='y', alpha=0.3)

plt.show()
```

#### Interpretation

- Each toss represents one random experiment.
- Since the coin is fair, both outcomes occur with nearly equal frequency.
- As the number of experiments increases, the observed frequencies approach the theoretical probability of 50% Heads and 50% Tails (Law of Large Numbers).

### 8. Machine Learning Connection



Machine Learning models are built to predict the outcomes of random experiments.

Examples:

| Machine Learning Problem | Possible Outcomes |
|--------------------------|-------------------|
| Spam Detection | Spam / Not Spam |
| Disease Prediction | Disease / Healthy |
| Customer Churn | Churn / No Churn |
| Loan Approval | Approve / Reject |
| Image Classification | Cat / Dog |

Every prediction made by a machine learning model begins with a random experiment, where the exact outcome is unknown but the possible outcomes are predefined.

The model learns patterns from historical data and estimates the probability of each possible outcome.
