## Probability Rules

### 1. Introduction



**Probability** is a numerical measure of the likelihood that an event will occur.

It quantifies uncertainty by assigning a value between **0 and 1** to an event.

- **0** → Impossible Event
- **1** → Certain Event
- **0 < P(E) < 1** → Event may or may not occur

The greater the probability, the more likely the event is to occur.

Probability helps us make decisions under uncertainty and is the foundation of statistics, data science, and machine learning.


### 2. Intuition



Imagine rolling a fair six-sided die.

Can you predict the exact number?

No.

Can you estimate the chance of getting a **4**?

Yes.

Since there are six equally likely outcomes and only one favorable outcome,

the probability is

$$
\frac16
$$

Similarly,

- Weather forecasts estimate the probability of rain.
- Banks estimate the probability of loan defaults.
- Machine learning models estimate the probability of customer churn.

Probability allows us to measure uncertainty rather than make absolute predictions.


### 3. Mathematical Perspective


#### 📐 Formula

For equally likely outcomes,

$$
P(E)=\frac{\text{Number of Favorable Outcomes}}{\text{Total Number of Possible Outcomes}}
$$

where

- **P(E)** = Probability of Event E
- **Favorable Outcomes** = Outcomes satisfying the event
- **Total Outcomes** = Total outcomes in the sample sp#ace

---

### Example

Rolling a die

Sample Space

$$
S=\{1,2,3,4,5,6\}
$$

Event

Getting an even number

$$
E=\{2,4,6\}
$$

Therefore,

$$
P(E)=\frac36=\frac12
$$

## Probability Rules


Probability rules help us calculate the probability of events without listing every possible outcome. These rules simplify probability calculations and form the foundation for advanced topics such as Conditional Probability and Bayes' Theorem.

The three fundamental probability rules are:

1. Complement Rule
2. Addition Rule
3. Multiplication Rule

### Complement Rule


#### 📖 Concept

The **Complement Rule** states that the probability of an event **not occurring** is equal to **1 minus the probability of the event occurring**.

The complement of an event **A** is denoted by:

$$
A^c
$$

or

$$
A'
$$

---

#### 🧠 Intuition

Suppose there are only two possibilities:

- Rain
- No Rain

If the probability of rain is

$$
0.7
$$

then the probability of **not raining** must be

$$
0.3
$$

because together they make up the entire sample space.

---

#### 📐 Formula

$$
P(A^c)=1-P(A)
$$

Where:

- \(P(A)\) = Probability that event **A** occurs
- \(P(A^c)\) = Probability that event **A** does **not** occur

---

#### 💡 Quick Tip

| Situation | Formula |
|-----------|---------|
| Probability of an event **not occurring** | \(P(A^c)=1-P(A)\) |
| **NOT** keyword | Use the **Complement Rule** |\(P(A^c)\) = Probability that event A does not occur

### Python

```python
# Probability of Rain
P_rain = 0.7

P_no_rain = 1 - P_rain

print(P_no_rain)
```

### 2. Addition Rule

#### 📖 Concept

The **Addition Rule** is used to calculate the probability that **at least one of two events occurs**.

The keyword is **"OR"**.

It is commonly used when finding the probability of the **union of two events**.

---

#### 🧠 Intuition

Suppose you roll a fair die.

Let:

- **A** = Getting an **even number**
- **B** = Getting a **number greater than 4**

Instead of calculating these events separately, we ask:

> **What is the probability of getting an even number OR a number greater than 4?**

Since we are interested in **either event (or both)** occurring, we use the **Addition Rule**.

---

#### 📐 Formula

##### 1. General Addition Rule

For any two events:

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B)
$$

**Why do we subtract \(P(A \cap B)\)?**

The outcomes common to both events are counted **twice** when we add \(P(A)\) and \(P(B)\). Subtracting the intersection removes the duplicate count.

---

##### 2. Mutually Exclusive Events

If two events **cannot occur together**, then:

$$
A \cap B = \varnothing
$$

Therefore,

$$
P(A \cup B) = P(A) + P(B)
$$

**Examples:**
- Rolling a die: Getting **2** or **5**
- Tossing a coin once: Getting **Heads** or **Tails**

---

#### 💡 Quick Tip

| Situation | Formula |
|-----------|---------|
| **OR (General Case)** | \(P(A \cup B) = P(A) + P(B) - P(A \cap B)\) |
| **OR (Mutually Exclusive)** | \(P(A \cup B) = P(A) + P(B)\) |
| **Keyword** | **OR → Addition Rule** |

```python
P_even = 3/6
P_greater4 = 2/6
P_intersection = 1/6

P_union = P_even + P_greater4 - P_intersection

print(P_union)
```

## 3. Multiplication Rule

#### Concept

The **Multiplication Rule** is used to calculate the probability that **two or more events occur together**.

The keyword is **"AND"**.

Whenever a probability question contains **AND**, use the Multiplication Rule.

---

#### Intuition

Suppose you toss a fair coin **twice**.

What is the probability of getting **Heads AND Heads**?

Both events must occur together:

- First toss = Heads
- Second toss = Heads

Since both events must happen together, we multiply their probabilities.

---

#### Formula

##### 1. Independent Events

If the occurrence of one event **does not affect** the other event:

$$
P(A \cap B) = P(A) \times P(B)
$$

**Examples:**
- Tossing two coins
- Rolling two dice
- Drawing cards **with replacement**

---

##### 2. Dependent Events

If the occurrence of one event **affects** the probability of the next event:

$$
P(A \cap B) = P(A) \times P(B \mid A)
$$

Where:

- **\(P(B \mid A)\)** = Probability of **B given that A has already occurred**

**Examples:**
- Drawing cards **without replacement**
- Choosing students one after another without replacement

---

#### Quick Tip

| Keyword | Formula |
|----------|---------|
| **AND (Independent)** | \(P(A \cap B) = P(A) \times P(B)\) |
| **AND (Dependent)** | \(P(A \cap B) = P(A) \times P(B \mid A)\) |P(B)\)
> - **Dependent:** \(P(A \cap B) = P(A) \times P(B \mid A)\) \mid A)\)

```python
P_heads = 0.5

P_HH = P_heads * P_heads

print(P_HH)
```

```python
P_first_ace = 4/52

P_second_ace = 3/51

P_two_aces = P_first_ace * P_second_ace

print(P_two_aces)
```
