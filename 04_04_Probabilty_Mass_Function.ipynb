## Probability Mass Function

### 1. Introduction

A **Probability Mass Function (PMF)** gives the probability of each possible value of a **discrete random variable**.

In simple terms, PMF answers:

> **"What is the probability that the discrete random variable takes a particular value?"**

For example, if \(X\) represents the result of rolling a fair die:

$$
P(X=1)=\frac{1}{6}
$$

$$
P(X=2)=\frac{1}{6}
$$

and so on.

The PMF provides these probabilities for **all possible values of the discrete random variable**.


### 2. Intuition


Think of a PMF as a **lookup table for probabilities**.

Suppose:

$$
X=\text{number obtained when rolling a die}
$$

You can ask:

$$
P(X=4)
$$

The PMF gives the answer:

$$
P(X=4)=\frac{1}{6}
$$

Similarly:

$$
P(X=1)=\frac{1}{6}
$$

$$
P(X=2)=\frac{1}{6}
$$

and so on.

Therefore:

> **PMF tells us the probability associated with each specific value of a discrete random variable.**

### 3. Formula


For a discrete random variable \(X\), the PMF is written as:

$$
p(x)=P(X=x)
$$

This means:

> \(p(x)\) is the probability that the random variable \(X\) takes the value \(x\).

For example:

$$
p(4)=P(X=4)
$$

If \(X\) represents the result of a fair die:

$$
p(4)=\frac{1}{6}
$$

### 4. Conditions

A valid PMF must satisfy two conditions.

#### 1. Every probability must be between 0 and 1

$$
0\leq p(x)\leq1
$$

A probability cannot be negative or greater than 1.

#### 2. The probabilities must add up to 1

$$
\sum_x p(x)=1
$$

This means that the total probability across all possible values must equal 1.

#### Example

For a fair die:

$$
\frac16+\frac16+\frac16+\frac16+\frac16+\frac16=1
$$

Therefore, the probabilities form a valid PMF.

### 5. Example: PMF of a Fair Die


Let:

$$
X=\text{result of rolling a fair six-sided die}
$$

The possible values are:

$$
X\in\{1,2,3,4,5,6\}
$$

Since the die is fair, each value has probability:

$$
P(X=x)=\frac16
$$

The PMF is:

| \(x\) | \(P(X=x)\) |
|---:|---:|
| 1 | \(1/6\) |
| 2 | \(1/6\) |
| 3 | \(1/6\) |
| 4 | \(1/6\) |
| 5 | \(1/6\) |
| 6 | \(1/6\) |

For example:

$$
P(X=3)=\frac16
$$

This means there is a \(1/6\) probability of rolling a 3.

### 6. Python Implementation




#### dice

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 6])

pmf = np.array([1/6] * 6)

print("Values:", x)
print("PMF:", pmf)
print("Total probability:", pmf.sum())
```

### 7. Find a specific probability

Suppose we want to find:

$$
P(X=4)
$$

We can look up the probability corresponding to \(X=4\) in the PMF.

```python
probability_of_4 = pmf[x == 4][0]

print("P(X = 4):", probability_of_4)
```

### 8. Visualization

Because a PMF is used for a **discrete random variable**, a bar chart is a suitable visualization.

Each bar represents the probability of one possible value.

For a fair die, all bars have the same height because all outcomes are equally likely.

```python
import matplotlib.pyplot as plt

plt.bar(x, pmf)

plt.xlabel("Value of X")
plt.ylabel("P(X = x)")
plt.title("PMF of a Fair Die")

plt.xticks(x)

plt.show()
```

### 9. Unequal PMF

Consider a discrete random variable \(X\) with the following probabilities:

| \(X\) | \(P(X=x)\) |
|---:|---:|
| 0 | 0.1 |
| 1 | 0.2 |
| 2 | 0.5 |
| 3 | 0.2 |

The probabilities add up to:

$$
0.1+0.2+0.5+0.2=1
$$

Therefore, this is a valid PMF.

The most likely value is:

$$
X=2
$$

because:

$$
P(X=2)=0.5
$$

A PMF simply tells us how probability is distributed across the possible values. The probabilities do not have to be equal.

```python
x = np.array([0, 1, 2, 3])
pmf = np.array([0.1, 0.2, 0.5, 0.2])

plt.bar(x, pmf)

plt.xlabel("Value of X")
plt.ylabel("P(X = x)")
plt.title("PMF with Unequal Probabilities")

plt.show()
```

### 10. Connection to Expected Value

Once we have the PMF of a discrete random variable, we can calculate its expected value.

The formula is:

$$
E[X]=\sum_x xp(x)
$$

where:

- \(x\) = possible value
- \(p(x)\) = probability of that value

For example, consider:

| \(X\) | \(P(X=x)\) |
|---:|---:|
| 0 | 0.1 |
| 1 | 0.2 |
| 2 | 0.5 |
| 3 | 0.2 |

Then:

$$
E[X]
=
0(0.1)+1(0.2)+2(0.5)+3(0.2)
$$

$$
=0+0.2+1+0.6
$$

$$
=1.8
$$

Therefore:

$$
E[X]=1.8
$$

This shows that the PMF provides the probabilities needed to calculate the expected value.

```python
expected_value = np.sum(x * pmf)

print("Expected Value:", expected_value)
```

### 11.  Probability Distribution vs PMF

A **probability distribution** describes how probabilities are assigned to the possible values of a random variable.

A **PMF** is the mathematical function used to describe a probability distribution when the random variable is discrete.

Therefore:

$$
\boxed{\text{Discrete Random Variable} \rightarrow \text{PMF}}
$$

For example:

$$
X=\text{number of customers}
$$

The PMF can tell us:

$$
P(X=0),P(X=1),P(X=2),P(X=3),\ldots
$$

So PMF gives us the probability associated with each possible value of a discrete random variable

### 14. Real Dataset Example — Titanic


The Titanic dataset contains the variable `SibSp`, which represents the number of siblings or spouses a passenger had aboard the Titanic.

This is a **discrete variable** because it represents a count.

Let:

$$
X=\text{Number of siblings/spouses aboard}
$$

We can calculate the empirical PMF by finding the proportion of passengers for each value of \(X\).

For example:

$$
P(X=0)
$$

represents the proportion of passengers who had zero siblings or spouses aboard.nger age.

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\WIN 10\Documents\stats and probability\Titanic-Dataset.csv")
pmf_sibsp = df["SibSp"].value_counts(normalize=True).sort_index()

print(pmf_sibsp)
```

### 15. Visualization

We can visualize the empirical PMF of `SibSp` using a bar chart.

Each bar represents the observed probability of a particular number of siblings or spouses aboard.

```python
plt.bar(pmf_sibsp.index, pmf_sibsp.values)

plt.xlabel("Number of Siblings/Spouses (SibSp)")
plt.ylabel("Probability")
plt.title("Empirical PMF of SibSp")

plt.show()
```

The resulting PMF tells us how frequently each value of `SibSp` occurs in the Titanic dataset.

For example, if:

$$
P(X=0)=0.68
$$

this would mean that approximately 68% of the passengers in the dataset had zero siblings or spouses aboard.

This is called an **empirical PMF** because the probabilities are calculated from observed data rather than from a theoretical probability model.
