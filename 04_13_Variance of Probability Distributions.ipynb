## Variance of a Probability Distribution

### 1. Introduction

Variance measures how much the values of a random variable are **spread out around their Expected Value**.

In simple words:

> **Expected Value tells us where the distribution is centered.**

> **Variance tells us how much the values vary around that center.**

A small variance means the values are concentrated close to the expected value.

A large variance means the values are more spread out.

### 2. Intuition

Consider two probability distributions with the same Expected Value:

#### Distribution A

Values are mostly close to the mean.

```text
     • • • • •
   •           •
─────────μ─────────


#### Distribution B

Values are spread far away from the mean.

```text
•       •       •       •
────────────μ────────────


Large spread → Large variance

Therefore:

Expected Value → Center

Variance → Spread


### 3. Variance Formula


For a discrete random variable:

$$
Var(X) = E[(X-\mu)^2]
$$

where:

- `X` = possible value
- `μ` = Expected Value
- `(X - μ)` = deviation from the mean

We square the deviations so that positive and negative deviations do not cancel each other.

An equivalent formula is:

$$
Var(X)=E[X^2]-(E[X])^2
$$

Both formulas give the same variance.

### 4.Simple Example

Consider:

| X | P(X) |
|---|---:|
| 1 | 0.25 |
| 2 | 0.50 |
| 3 | 0.25 |

First calculate the Expected Value:

$$
E[X] = (1)(0.25)+(2)(0.50)+(3)(0.25)
$$

$$
E[X]=2
$$

Now calculate the variance:

$$
Var(X)=\sum (x-\mu)^2P(X=x)
$$

$$
=(1-2)^2(0.25)+(2-2)^2(0.50)+(3-2)^2(0.25)
$$

$$
=0.25+0+0.25
$$

$$
Var(X)=0.5
$$

So the variance is **0.5**.

````python
import numpy as np

x = np.array([1, 2, 3])
probabilities = np.array([0.25, 0.50, 0.25])

mean = np.sum(x * probabilities)

variance = np.sum((x - mean) ** 2 * probabilities)

print("Expected Value:", mean)
print("Variance:", variance)
````


### 5. Variance of Common Distributions


Different probability distributions have different variance formulas.

#### Bernoulli

$$
Var(X)=p(1-p)
$$

#### Binomial

$$
Var(X)=np(1-p)
$$

#### Poisson

$$
Var(X)=\lambda
$$

#### Uniform

For a continuous Uniform Distribution between `a` and `b`:

$$
Var(X)=\frac{(b-a)^2}{12}
$$

#### Normal

For a Normal Distribution:

$$
Var(X)=\sigma^2
$$

where `σ` is the standard deviation.

### 4. Example 


 #### a. Binomial Distribution

Suppose an advertisement is shown to **100 customers**.

The probability that a customer clicks the advertisement is **0.2**.

Therefore:

- `n = 100`
- `p = 0.2`

For a Binomial Distribution:

$$
Var(X)=np(1-p)
$$

Substituting:

$$
Var(X)=100(0.2)(1-0.2)
$$

$$
Var(X)=100(0.2)(0.8)
$$

$$
Var(X)=16
$$

So the **variance is 16**.

The standard deviation is:

$$
\sigma=\sqrt{16}=4
$$

This means the number of clicks typically varies around the expected value.

The expected number of clicks is:

$$
E[X]=np=100(0.2)=20
$$

So:

**Expected clicks = 20**

**Variance = 16**

**Standard deviation = 4**

````python
from scipy.stats import binom

n = 100
p = 0.2

mean = binom.mean(n, p)
variance = binom.var(n, p)
std = binom.std(n, p)

print("Expected Value:", mean)
print("Variance:", variance)
print("Standard Deviation:", std)
````

#### b. Poisson Distribution

Suppose a call center receives an average of **8 calls per hour**.

For a Poisson Distribution:

$$
Var(X)=\lambda
$$

Therefore:

$$
Var(X)=8
$$

The variance of the number of calls per hour is **8**.

Notice an important property of the Poisson Distribution:

$$
Mean = Variance = \lambda
$$

So here:

**Mean = 8**

**Variance = 8**

````python
from scipy.stats import poisson

lambda_value = 8

variance = poisson.var(lambda_value)

print("Variance:", variance)
````

#### c. Normal Distribution

Suppose:

- Mean = 170 cm
- Standard deviation = 10 cm

For a Normal Distribution:

$$
Var(X)=\sigma^2
$$

Therefore:

$$
Var(X)=10^2=100
$$

So:

**Variance = 100 cm²**

**Standard deviation = 10 cm**

This shows an important relationship:

$$
Standard\ Deviation = \sqrt{Variance}
$$

and

$$
Variance = (Standard\ Deviation)^2
$$

````python
from scipy.stats import norm

mean = 170
std = 10

expected_value = norm.mean(loc=mean, scale=std)
variance = norm.var(loc=mean, scale=std)
standard_deviation = norm.std(loc=mean, scale=std)

print("Expected Value:", expected_value)
print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
````

### Distribution Comparison

| Distribution | Expected Value | Variance |
|---|---:|---:|
| Bernoulli | `p` | `p(1-p)` |
| Binomial | `np` | `np(1-p)` |
| Poisson | `λ` | `λ` |
| Uniform | `(a+b)/2` | `(b-a)²/12` |
| Normal | `μ` | `σ²` |

The important idea is:

**Expected Value → Center**

**Variance → Spread**

**Standard Deviation → Spread in original units**
