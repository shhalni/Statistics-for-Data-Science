```python
# Merge datasets using Order ID
merged = pd.merge(order_list, order_details, on='Order ID')
```

```python
merged.head()
```

```python
# Total Sales by Category
category_sales = merged.groupby('Category')['Amount'].sum()
```

```python
category_sales
```

```python
# Average Profit per Order
avg_profit = merged.groupby('Category').agg(Average_Profit_Per_Order=("Profit", "mean"))
```

```python
avg_profit
```

```python
# Profit Margin
profit_stats = merged.groupby("Category").agg(Total_Sales=("Amount", "sum"),Total_Profit=("Profit", "sum"))
```

```python
profit_stats
```

```python
profit_stats['Profit_Margin (%)'] = ( profit_stats['Total_Profit'] / profit_stats['Total_Sales']) * 100
```

```python
profit_stats
```

```python
category_analysis = profit_stats.join(avg_profit)
```

```python
category_analysis
```

```python
best = category_analysis.sort_values(
    by=["Profit_Margin (%)", "Average_Profit_Per_Order"],
    ascending=False
)

# Worst Category
worst = category_analysis.sort_values(
    by=["Profit_Margin (%)", "Average_Profit_Per_Order"]
)

print("\nTop Performing Category")
print(best.head(1))

print("\nUnderperforming Category")
print(worst.head(1))
```
